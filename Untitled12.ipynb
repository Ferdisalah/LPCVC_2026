```python
# ============================================================
# HOME CREDIT
# 20 CATBOOST MODELS + 20 ROOT DECISION VECTORS
#
# GROUP A: 10 models, original class distribution
#          2 repetitions x 5-fold StratifiedKFold
#          each model trains on 80% of the data
#
# GROUP B: 10 models, negative:positive ratio = 3:1
#          disjoint positive samples and disjoint negative samples
#
# OUTPUT:
#   20 CatBoost models (.cbm)
#   20 JSON model exports
#   20 binary root vectors, each length 900
# ============================================================

import os
import sys
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold

try:
    from catboost import CatBoostClassifier, Pool
except ImportError:
    os.system(f"{sys.executable} -m pip -q install catboost")
    from catboost import CatBoostClassifier, Pool


# ============================================================
# 1. CONFIGURATION
# ============================================================

TRAIN_PATH = "/content/drive/MyDrive/Competition/input/train.csv"

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/Competition/output/"
    "CATBOOST_20_ROOT_VECTORS"
)

MODEL_DIR = OUTPUT_DIR / "models"
VECTOR_DIR = OUTPUT_DIR / "vectors"
INDIVIDUAL_DIR = VECTOR_DIR / "individual"

for folder in [OUTPUT_DIR, MODEL_DIR, VECTOR_DIR, INDIVIDUAL_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

TARGET = "TARGET"
ID_COL = "SK_ID_CURR"

SEED = 42

N_MODELS = 20
N_MODELS_A = 10
N_MODELS_B = 10

N_SPLITS = 5
N_REPEATS_A = 2

# Exactly 900 trees in each model.
CB_ITERATIONS = 900
CB_DEPTH = 7
CB_LEARNING_RATE = 0.04
CB_L2_LEAF_REG = 10.0
CB_VERBOSE = 200

CAT_FEATURES = [
    "NAME_CONTRACT_TYPE",
    "CODE_GENDER",
    "NAME_INCOME_TYPE",
    "NAME_EDUCATION_TYPE",
    "NAME_FAMILY_STATUS",
    "NAME_HOUSING_TYPE",
    "OCCUPATION_TYPE",
    "ORGANIZATION_TYPE",
]

# CatBoost JSON split types.
NUMERICAL_TYPES = {
    "FloatFeature"
}

CATEGORICAL_TYPES = {
    "OneHotFeature",
    "OnlineCtr",
}

random.seed(SEED)
np.random.seed(SEED)

warnings.filterwarnings("ignore", category=FutureWarning)


# ============================================================
# 2. LOAD DATA
# ============================================================

print("\n[1/6] Loading data...")

train = pd.read_csv(TRAIN_PATH)

required = [
    TARGET,
    ID_COL,
    "AMT_CREDIT",
    "AMT_ANNUITY",
] + CAT_FEATURES

missing = [c for c in required if c not in train.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

if train[ID_COL].duplicated().any():
    raise ValueError("Duplicate SK_ID_CURR values detected.")

y = train[TARGET].astype(np.int64).to_numpy()
train_ids = train[ID_COL].to_numpy()

if set(np.unique(y)) != {0, 1}:
    raise ValueError("TARGET must contain only 0 and 1.")

print("Train shape:", train.shape)
print("Non-default:", int((y == 0).sum()))
print("Default:", int((y == 1).sum()))
print("Default rate:", round(y.mean() * 100, 4), "%")


# ============================================================
# 3. PREPARE CATBOOST FEATURES
# ============================================================

print("\n[2/6] Preparing CatBoost features...")

# All original predictors except TARGET and ID.
cb_feature_cols = [
    c for c in train.columns
    if c not in [TARGET, ID_COL]
]

X_cb = train[cb_feature_cols].copy()

# Add CREDIT_TERM exactly as in the supplied script.
credit = pd.to_numeric(
    train["AMT_CREDIT"],
    errors="coerce"
)

annuity = pd.to_numeric(
    train["AMT_ANNUITY"],
    errors="coerce"
).replace(0, np.nan)

X_cb["CREDIT_TERM"] = credit / annuity

X_cb["CREDIT_TERM"] = X_cb["CREDIT_TERM"].replace(
    [np.inf, -np.inf],
    np.nan
)

# Handle categorical missing values.
for col in CAT_FEATURES:
    X_cb[col] = (
        X_cb[col]
        .where(X_cb[col].notna(), "__MISSING__")
        .astype(str)
    )

assert TARGET not in X_cb.columns
assert ID_COL not in X_cb.columns

print("CatBoost input shape:", X_cb.shape)
print("Categorical features:", len(CAT_FEATURES))
print("Added feature: CREDIT_TERM")


# ============================================================
# 4. CREATE 20 TRAINING SAMPLES
# ============================================================

print("\n[3/6] Creating model training samples...")

model_specs = []


# ------------------------------------------------------------
# GROUP A
# 10 models with original class distribution.
# 2 repetitions x 5 folds.
# Each model trains on 80% of the original dataset.
# ------------------------------------------------------------

for repeat in range(N_REPEATS_A):

    splitter = StratifiedKFold(
        n_splits=N_SPLITS,
        shuffle=True,
        random_state=SEED + repeat,
    )

    for fold, (fit_idx, valid_idx) in enumerate(
        splitter.split(X_cb, y)
    ):

        model_name = (
            f"A_original_r{repeat + 1}_f{fold + 1}"
        )

        model_specs.append({
            "model_name": model_name,
            "group": "A_original",
            "fit_idx": np.asarray(fit_idx, dtype=np.int64),
            "seed": SEED + 100 * repeat + fold,
            "sampling": "Original distribution, 80% fit split",
        })


# ------------------------------------------------------------
# GROUP B
# 10 models with:
# Non-default : Default = 3 : 1
#
# Positive rows are divided into 10 disjoint groups.
# Negative rows are also divided into disjoint groups.
#
# No training row is shared between two GROUP B models.
# ------------------------------------------------------------

positive_idx = np.flatnonzero(y == 1)
negative_idx = np.flatnonzero(y == 0)

rng = np.random.default_rng(SEED + 5000)

positive_idx = rng.permutation(positive_idx)
negative_idx = rng.permutation(negative_idx)

positive_chunks = np.array_split(
    positive_idx,
    N_MODELS_B
)

negative_cursor = 0

for model_no, positive_chunk in enumerate(
    positive_chunks,
    start=1
):

    n_positive = len(positive_chunk)
    n_negative = 3 * n_positive

    next_cursor = negative_cursor + n_negative

    if next_cursor > len(negative_idx):
        raise ValueError(
            "Not enough negative samples for disjoint 3:1 sampling."
        )

    negative_chunk = negative_idx[
        negative_cursor:next_cursor
    ]

    negative_cursor = next_cursor

    fit_idx = np.concatenate([
        positive_chunk,
        negative_chunk
    ])

    # Shuffle rows inside this model's training sample.
    model_rng = np.random.default_rng(
        SEED + 6000 + model_no
    )

    fit_idx = model_rng.permutation(fit_idx)

    model_specs.append({
        "model_name": f"B_ratio3to1_{model_no:02d}",
        "group": "B_ratio3to1",
        "fit_idx": fit_idx.astype(np.int64),
        "seed": SEED + 2000 + model_no,
        "sampling": "Disjoint samples with negative:positive = 3:1",
    })


assert len(model_specs) == 20

print("Total model specifications:", len(model_specs))


# ============================================================
# 5. VERIFY TRAINING SAMPLES AND SAVE MANIFEST
# ============================================================

print("\n[4/6] Verifying samples...")

manifest_rows = []
summary_rows = []

sample_sets = {}

for spec in model_specs:

    idx = spec["fit_idx"]
    subset_y = y[idx]

    n_positive = int((subset_y == 1).sum())
    n_negative = int((subset_y == 0).sum())

    if spec["group"] == "B_ratio3to1":

        assert n_negative == 3 * n_positive

    row_ids = train_ids[idx]

    sample_sets[spec["model_name"]] = set(
        row_ids.tolist()
    )

    summary_rows.append({
        "model_name": spec["model_name"],
        "group": spec["group"],
        "train_rows": len(idx),
        "non_default_rows": n_negative,
        "default_rows": n_positive,
        "negative_positive_ratio": n_negative / n_positive,
        "default_rate": n_positive / len(idx),
        "seed": spec["seed"],
    })

    for row_idx in idx:

        manifest_rows.append({
            "model_name": spec["model_name"],
            "group": spec["group"],
            ID_COL: train_ids[row_idx],
            TARGET: int(y[row_idx]),
        })


summary_df = pd.DataFrame(summary_rows)

summary_df.to_csv(
    OUTPUT_DIR / "model_training_summary.csv",
    index=False
)

manifest_df = pd.DataFrame(manifest_rows)

manifest_df.to_csv(
    OUTPUT_DIR / "training_sample_manifest.csv",
    index=False
)

print(summary_df.to_string(index=False))


# ------------------------------------------------------------
# Verify zero overlap inside GROUP B.
# ------------------------------------------------------------

group_b_names = [
    s["model_name"]
    for s in model_specs
    if s["group"] == "B_ratio3to1"
]

for i in range(len(group_b_names)):

    for j in range(i + 1, len(group_b_names)):

        set_i = sample_sets[group_b_names[i]]
        set_j = sample_sets[group_b_names[j]]

        overlap = set_i.intersection(set_j)

        if len(overlap) > 0:
            raise RuntimeError(
                f"Overlap found between "
                f"{group_b_names[i]} and {group_b_names[j]}"
            )

print("\nGROUP B: zero training-row overlap verified.")
```

```python

# ============================================================
# 6. PAIRWISE TRAINING OVERLAP REPORT
# ============================================================

print("\nCreating pairwise overlap report...")

overlap_rows = []

for i in range(len(model_specs)):

    name_i = model_specs[i]["model_name"]
    set_i = sample_sets[name_i]

    for j in range(i + 1, len(model_specs)):

        name_j = model_specs[j]["model_name"]
        set_j = sample_sets[name_j]

        intersection = len(set_i.intersection(set_j))
        union = len(set_i.union(set_j))

        overlap_rows.append({
            "model_1": name_i,
            "model_2": name_j,
            "group_1": model_specs[i]["group"],
            "group_2": model_specs[j]["group"],
            "overlap_rows": intersection,
            "overlap_fraction_smaller": (
                intersection / min(len(set_i), len(set_j))
            ),
            "jaccard_overlap": intersection / union if union > 0 else 0.0,
        })

pd.DataFrame(overlap_rows).to_csv(
    OUTPUT_DIR / "pairwise_train_overlap.csv",
    index=False
)


# ============================================================
# 7. ROOT SPLIT EXTRACTION
# ============================================================

def extract_root_vector(json_path):

    with open(json_path, "r", encoding="utf-8") as f:
        model_json = json.load(f)

    trees = model_json.get("oblivious_trees")

    if trees is None:
        raise RuntimeError(
            f"Missing oblivious_trees in {json_path}"
        )

    if len(trees) != CB_ITERATIONS:
        raise RuntimeError(
            f"Expected {CB_ITERATIONS} trees, "
            f"found {len(trees)}"
        )

    # Encoding:
    # 0 = Numerical root split
    # 1 = Categorical root split
    # 2 = No split

    vector = np.full(
        CB_ITERATIONS,
        2,
        dtype=np.uint8
    )

    audit = []

    for tree_idx, tree in enumerate(trees):

        splits = tree.get("splits") or []

        # ----------------------------------------------------
        # CASE 1: TREE WITHOUT SPLIT
        # ----------------------------------------------------

        if len(splits) == 0:

            vector[tree_idx] = 2

            audit.append({
                "tree_index": tree_idx + 1,
                "root_value": 2,
                "root_kind": "no_split",
                "root_split_type": None,
                "root_split_details": None,
                "has_root_split": False,
            })

            continue

        # ----------------------------------------------------
        # CASE 2: TREE HAS SPLITS
        # ----------------------------------------------------

        # CatBoost symmetric-tree JSON stores split levels
        # in reverse order. The LAST element is the root.

        root_split = splits[-1]

        split_type = root_split.get("split_type")

        if split_type in NUMERICAL_TYPES:

            value = 0
            kind = "numerical"

        elif split_type in CATEGORICAL_TYPES:

            value = 1
            kind = "categorical"

        else:

            raise RuntimeError(
                f"Unknown root split type: {split_type} "
                f"in tree {tree_idx + 1}, file: {json_path}"
            )

        vector[tree_idx] = value

        audit.append({
            "tree_index": tree_idx + 1,
            "root_value": value,
            "root_kind": kind,
            "root_split_type": split_type,
            "root_split_details": json.dumps(
                root_split,
                ensure_ascii=False
            ),
            "has_root_split": True,
        })

    # --------------------------------------------------------
    # FINAL VECTOR VALIDATION
    # --------------------------------------------------------

    assert vector.shape == (CB_ITERATIONS,)

    assert np.isin(vector, [0, 1, 2]).all()

    assert len(audit) == CB_ITERATIONS

    return vector, audit


# ============================================================
# 8. TRAIN THE 20 MODELS
# ============================================================

print("\n" + "=" * 75)
print("TRAINING 20 CATBOOST MODELS")
print("=" * 75)

all_vectors = []
all_audits = []
vector_summary = []

for model_number, spec in enumerate(
    model_specs,
    start=1
):

    model_name = spec["model_name"]
    fit_idx = spec["fit_idx"]

    print("\n" + "-" * 75)
    print(f"MODEL {model_number}/20: {model_name}")
    print("Group:", spec["group"])
    print("Training rows:", len(fit_idx))
    print("Non-default:", int((y[fit_idx] == 0).sum()))
    print("Default:", int((y[fit_idx] == 1).sum()))

    train_pool = Pool(
        X_cb.iloc[fit_idx],
        y[fit_idx],
        cat_features=CAT_FEATURES,
    )

    cb_model = CatBoostClassifier(
        iterations=CB_ITERATIONS,
        depth=CB_DEPTH,
        learning_rate=CB_LEARNING_RATE,
        l2_leaf_reg=CB_L2_LEAF_REG,
        loss_function="Logloss",
        eval_metric="AUC",
        random_seed=int(spec["seed"]),
        verbose=CB_VERBOSE,
        allow_writing_files=False,
        thread_count=-1,
        grow_policy="SymmetricTree",
    )

    # No validation-based early stopping.
    # Every model must contain exactly CB_ITERATIONS trees.

    cb_model.fit(train_pool)

    if cb_model.tree_count_ != CB_ITERATIONS:

        raise RuntimeError(
            f"{model_name}: expected {CB_ITERATIONS} trees, "
            f"got {cb_model.tree_count_}"
        )

    # --------------------------------------------------------
    # Save native CatBoost model.
    # --------------------------------------------------------

    model_path = MODEL_DIR / f"{model_name}.cbm"

    cb_model.save_model(
        str(model_path),
        format="cbm"
    )

    # --------------------------------------------------------
    # Export JSON for root split extraction.
    # --------------------------------------------------------

    json_path = MODEL_DIR / f"{model_name}.json"

    cb_model.save_model(
        str(json_path),
        format="json"
    )

    # --------------------------------------------------------
    # Extract root vectors.
    # 0 = Numerical
    # 1 = Categorical
    # 2 = No split
    # --------------------------------------------------------

    vector, audit = extract_root_vector(json_path)

    for row in audit:

        row["model_name"] = model_name
        row["group"] = spec["group"]

    all_vectors.append(vector)
    all_audits.extend(audit)

    # --------------------------------------------------------
    # Save one vector per model.
    # --------------------------------------------------------

    np.save(
        INDIVIDUAL_DIR / f"{model_name}_vector.npy",
        vector
    )

    vector_columns = [
        f"TREE_{i:03d}"
        for i in range(1, CB_ITERATIONS + 1)
    ]

    pd.DataFrame(
        [vector],
        columns=vector_columns
    ).to_csv(
        INDIVIDUAL_DIR / f"{model_name}_vector.csv",
        index=False
    )

    # --------------------------------------------------------
    # Count the three root types independently.
    # --------------------------------------------------------

    n_num = int(np.sum(vector == 0))
    n_cat = int(np.sum(vector == 1))
    n_no_split = int(np.sum(vector == 2))

    vector_summary.append({

        "model_name": model_name,
        "group": spec["group"],
        "train_rows": len(fit_idx),

        "numerical_root_count": n_num,
        "categorical_root_count": n_cat,
        "no_split_tree_count": n_no_split,

        "numerical_fraction": n_num / CB_ITERATIONS,
        "categorical_fraction": n_cat / CB_ITERATIONS,
        "no_split_fraction": n_no_split / CB_ITERATIONS,

    })

    print("Model saved:", model_path)
    print("Vector shape:", vector.shape)

    print("Numerical roots:", n_num)
    print("Categorical roots:", n_cat)
    print("Trees without splits:", n_no_split)

    print(
        "Total:",
        n_num + n_cat + n_no_split
    )

    del train_pool
    del cb_model


# ============================================================
# 9. SAVE THE 20 VECTORS
# ============================================================

print("\nSaving the 20 root vectors...")

root_vectors = np.stack(
    all_vectors,
    axis=0
).astype(np.uint8)

if root_vectors.shape != (20, CB_ITERATIONS):

    raise RuntimeError(
        f"Expected (20, {CB_ITERATIONS}), "
        f"got {root_vectors.shape}"
    )

# Validate all three encodings.

if not np.isin(root_vectors, [0, 1, 2]).all():

    raise RuntimeError(
        "Invalid values found in root_vectors."
    )

# Main matrix: 20 models x 900 trees.

np.save(
    VECTOR_DIR / "root_vectors.npy",
    root_vectors
)

# Readable CSV with one row per model.

vector_columns = [
    f"TREE_{i:03d}"
    for i in range(1, CB_ITERATIONS + 1)
]

root_df = pd.DataFrame(
    root_vectors,
    columns=vector_columns
)

root_df.insert(
    0,
    "model_name",
    [s["model_name"] for s in model_specs]
)

root_df.insert(
    1,
    "group",
    [s["group"] for s in model_specs]
)

root_df.to_csv(
    VECTOR_DIR / "root_vectors.csv",
    index=False
)


# ============================================================
# 10. SAVE TREE-LEVEL AUDIT
# ============================================================

audit_df = pd.DataFrame(all_audits)

audit_df.to_csv(
    OUTPUT_DIR / "root_split_audit.csv",
    index=False
)


# ============================================================
# 11. SAVE VECTOR SUMMARY
# ============================================================

summary_vectors_df = pd.DataFrame(vector_summary)

summary_vectors_df.to_csv(
    VECTOR_DIR / "root_vector_summary.csv",
    index=False
)

print("\nRoot vector summary:")

print(
    summary_vectors_df[
        [
            "model_name",
            "group",
            "numerical_root_count",
            "categorical_root_count",
            "no_split_tree_count",
        ]
    ].to_string(index=False)
)


# ============================================================
# 12. SAVE MODEL ORDER
# ============================================================

order_df = pd.DataFrame({

    "matrix_row": np.arange(1, 21),

    "model_name": [
        s["model_name"] for s in model_specs
    ],

    "group": [
        s["group"] for s in model_specs
    ],

})

order_df.to_csv(
    VECTOR_DIR / "root_vector_model_order.csv",
    index=False
)


# ============================================================
# 13. SAVE EXPERIMENT METADATA
# ============================================================

metadata = {

    "total_models": 20,
    "group_a_models": 10,
    "group_b_models": 10,

    "trees_per_model": CB_ITERATIONS,

    "root_vector_shape": [
        20,
        CB_ITERATIONS
    ],

    "encoding": {

        "0": "Numerical root split",

        "1": "Categorical root split",

        "2": "Tree without split",

    },

    "categorical_split_types": sorted(
        CATEGORICAL_TYPES
    ),

    "numerical_split_types": sorted(
        NUMERICAL_TYPES
    ),

    "catboost_parameters": {

        "iterations": CB_ITERATIONS,

        "depth": CB_DEPTH,

        "learning_rate": CB_LEARNING_RATE,

        "l2_leaf_reg": CB_L2_LEAF_REG,

        "loss_function": "Logloss",

        "eval_metric": "AUC",

        "grow_policy": "SymmetricTree",

        "class_weights": None,

    },

    "group_a_sampling": (

        "Two repetitions of 5-fold StratifiedKFold; "
        "10 models, each trained on 80%."

    ),

    "group_b_sampling": (

        "10 models with negative:positive ratio 3:1; "
        "positive and negative training rows are disjoint "
        "between group B models."

    ),

    "training_rows_group_b_unused_negatives": int(
        len(negative_idx) - negative_cursor
    ),

    "cross_group_overlap_allowed": True,

    "seed": SEED,

    "model_order": [
        s["model_name"] for s in model_specs
    ],

}

with open(
    OUTPUT_DIR / "experiment_metadata.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2,
        ensure_ascii=False
    )


# ============================================================
# 14. FINAL VALIDATION
# ============================================================

print("\nFinal validation...")

assert len(list(MODEL_DIR.glob("*.cbm"))) == 20

assert len(list(MODEL_DIR.glob("*.json"))) == 20

assert len(
    list(INDIVIDUAL_DIR.glob("*_vector.npy"))
) == 20

assert root_vectors.shape == (20, CB_ITERATIONS)

assert np.isin(root_vectors, [0, 1, 2]).all()

assert len(all_audits) == 20 * CB_ITERATIONS

# Verify that every model has exactly CB_ITERATIONS entries.

for i in range(20):

    assert len(root_vectors[i]) == CB_ITERATIONS

    assert (
        np.sum(root_vectors[i] == 0)
        + np.sum(root_vectors[i] == 1)
        + np.sum(root_vectors[i] == 2)
    ) == CB_ITERATIONS


print("\n" + "=" * 75)
print("EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 75)

print("CatBoost models:", len(list(MODEL_DIR.glob("*.cbm"))))

print("JSON models:", len(list(MODEL_DIR.glob("*.json"))))

print(
    "Individual vectors:",
    len(list(INDIVIDUAL_DIR.glob("*_vector.npy")))
)

print("Combined vector shape:", root_vectors.shape)

print("\nEncoding:")
print("0 = Numerical")
print("1 = Categorical")
print("2 = No split")

print("\nTotal numerical roots:",
      int(np.sum(root_vectors == 0)))

print("Total categorical roots:",
      int(np.sum(root_vectors == 1)))

print("Total trees without splits:",
      int(np.sum(root_vectors == 2)))

print("\nMain output directory:")
print(OUTPUT_DIR)

print("\nVector matrix:")
print(VECTOR_DIR / "root_vectors.npy")

print("\nReadable vectors:")
print(VECTOR_DIR / "root_vectors.csv")

print("\nRoot split audit:")
print(OUTPUT_DIR / "root_split_audit.csv")

print("\nTraining sample manifest:")
print(OUTPUT_DIR / "training_sample_manifest.csv")

print("\nPairwise overlap:")
print(OUTPUT_DIR / "pairwise_train_overlap.csv")
```

```python

# ============================================================
# 15. TEST PREDICTIONS + 20 INDIVIDUAL SUBMISSIONS + ENSEMBLES
# ============================================================

import re
import numpy as np
import pandas as pd
from pathlib import Path
from catboost import CatBoostClassifier

print("\n" + "=" * 75)
print("GENERATING TEST SUBMISSIONS FOR ALL 20 CATBOOST MODELS")
print("=" * 75)


# ============================================================
# 1. CONFIGURATION
# ============================================================

# Exact paths from the training experiment
MODEL_ROOT = Path(
    "/content/drive/MyDrive/Competition/output/"
    "CATBOOST_20_ROOT_VECTORS"
)

MODELS_DIR = MODEL_ROOT / "models"

SUBMISSION_DIR = MODEL_ROOT / "test_submissions"
SUBMISSION_DIR.mkdir(parents=True, exist_ok=True)

ID_COL = "SK_ID_CURR"
TARGET_COL = "TARGET"

EXPECTED_MODELS = 20


# ============================================================
# 2. GET TEST DATA
# ============================================================

# Your original dataset is named "test", not "test_df".
# Copy it to test_df for use throughout this section.

if "test" in globals():

    test_df = test.copy()

elif "test_df" in globals():

    test_df = test_df.copy()

else:

    raise NameError(
        "Neither 'test' nor 'test_df' exists. "
        "Load the competition test.csv before running this section."
    )

if ID_COL not in test_df.columns:

    raise ValueError(
        f"Test dataset is missing {ID_COL}."
    )

if TARGET_COL in test_df.columns:

    raise ValueError(
        "The competition test dataset must not contain TARGET."
    )

print(f"Test rows: {len(test_df):,}")
print(f"Test columns: {len(test_df.columns)}")


# ============================================================
# 3. VALIDATE MODEL DIRECTORY
# ============================================================

if not MODELS_DIR.exists():

    raise FileNotFoundError(
        f"Models directory does not exist:\n{MODELS_DIR}"
    )

model_paths = list(MODELS_DIR.glob("*.cbm"))


# Natural sorting:
# A_original_r1_f1
# A_original_r1_f2
# ...
# B_ratio3to1_01
# ...

def natural_sort_key(path):

    return [
        int(part) if part.isdigit() else part.lower()
        for part in re.split(r"(\d+)", path.stem)
    ]


model_paths = sorted(
    model_paths,
    key=natural_sort_key
)


if len(model_paths) != EXPECTED_MODELS:

    raise RuntimeError(
        f"Expected {EXPECTED_MODELS} models, "
        f"but found {len(model_paths)}.\n"
        f"Directory: {MODELS_DIR}"
    )

print(f"\nFound {len(model_paths)} CatBoost models.")


# ============================================================
# 4. PREDICT WITH EACH MODEL
# ============================================================

all_predictions = {}

model_groups = {}

for model_idx, model_path in enumerate(
    model_paths,
    start=1
):

    model_name = model_path.stem

    print("\n" + "-" * 65)
    print(
        f"MODEL {model_idx}/{EXPECTED_MODELS}: "
        f"{model_name}"
    )

    # --------------------------------------------------------
    # Load model
    # --------------------------------------------------------

    model = CatBoostClassifier()
    model.load_model(str(model_path))

    # --------------------------------------------------------
    # Get exact training feature order
    # --------------------------------------------------------

    feature_names = model.feature_names_

    if not feature_names:

        raise RuntimeError(
            f"{model_name} has no stored feature names."
        )

    # --------------------------------------------------------
    # Check missing features
    # --------------------------------------------------------

    missing_features = [
        feature
        for feature in feature_names
        if feature not in test_df.columns
    ]

    if missing_features:

        raise ValueError(
            f"\nMissing features for {model_name}:\n"
            f"{missing_features}\n\n"
            "Make sure test data has undergone exactly "
            "the same feature engineering and preprocessing "
            "as the training data."
        )

    # --------------------------------------------------------
    # Select features in EXACT training order
    # --------------------------------------------------------

    X_test = test_df[feature_names]

    # --------------------------------------------------------
    # Find positive class TARGET = 1
    # --------------------------------------------------------

    classes = list(model.classes_)

    positive_class_indices = [
        i for i, cls in enumerate(classes)
        if cls == 1 or str(cls) == "1"
    ]

    if len(positive_class_indices) != 1:

        raise RuntimeError(
            f"Could not identify TARGET=1 in classes "
            f"for {model_name}: {classes}"
        )

    positive_class_idx = positive_class_indices[0]

    # --------------------------------------------------------
    # Predict probabilities
    # --------------------------------------------------------

    probabilities = model.predict_proba(
        X_test
    )[:, positive_class_idx]

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )

    if len(probabilities) != len(test_df):

        raise RuntimeError(
            f"Prediction length mismatch for {model_name}."
        )

    if not np.isfinite(probabilities).all():

        raise RuntimeError(
            f"Invalid predictions detected in {model_name}."
        )

    if probabilities.min() < 0 or probabilities.max() > 1:

        raise RuntimeError(
            f"Probabilities outside [0, 1] in {model_name}."
        )

    # Store predictions
    all_predictions[model_name] = probabilities

    # Identify model group
    if model_name.startswith("A_original_"):

        model_groups[model_name] = "A_original"

    elif model_name.startswith("B_ratio3to1_"):

        model_groups[model_name] = "B_ratio3to1"

    else:

        raise ValueError(
            f"Unknown model group: {model_name}"
        )

    # --------------------------------------------------------
    # SAVE INDIVIDUAL SUBMISSION
    # --------------------------------------------------------

    individual_submission = pd.DataFrame({

        ID_COL: test_df[ID_COL].values,

        TARGET_COL: probabilities

    })

    individual_path = (
        SUBMISSION_DIR /
        f"submission_{model_name}.csv"
    )

    individual_submission.to_csv(
        individual_path,
        index=False
    )

    print(f"Saved: {individual_path.name}")

    print(
        f"Mean: {probabilities.mean():.6f} | "
        f"Std: {probabilities.std():.6f} | "
        f"Min: {probabilities.min():.6f} | "
        f"Max: {probabilities.max():.6f}"
    )

    del model
    del X_test


# ============================================================
# 5. VALIDATE MODEL GROUPS
# ============================================================

group_A_models = [
    name for name, group in model_groups.items()
    if group == "A_original"
]

group_B_models = [
    name for name, group in model_groups.items()
    if group == "B_ratio3to1"
]

if len(group_A_models) != 10:

    raise RuntimeError(
        f"Expected 10 A_original models, "
        f"found {len(group_A_models)}."
    )

if len(group_B_models) != 10:

    raise RuntimeError(
        f"Expected 10 B_ratio3to1 models, "
        f"found {len(group_B_models)}."
    )


# ============================================================
# 6. BUILD COMPLETE PREDICTION MATRIX
# ============================================================

prediction_matrix = pd.DataFrame({

    ID_COL: test_df[ID_COL].values

})

for model_name, probabilities in all_predictions.items():

    prediction_matrix[model_name] = probabilities


# ============================================================
# 7. ENSEMBLE A: ORIGINAL DISTRIBUTION
# ============================================================

prediction_matrix["ENSEMBLE_A_ORIGINAL"] = (
    prediction_matrix[group_A_models].mean(axis=1)
)

submission_A = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_A_ORIGINAL"
    ].values

})

submission_A.to_csv(
    SUBMISSION_DIR / "submission_ensemble_A_original.csv",
    index=False
)


# ============================================================
# 8. ENSEMBLE B: RATIO 3:1
# ============================================================

prediction_matrix["ENSEMBLE_B_RATIO3TO1"] = (
    prediction_matrix[group_B_models].mean(axis=1)
)

submission_B = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_B_RATIO3TO1"
    ].values

})

submission_B.to_csv(
    SUBMISSION_DIR / "submission_ensemble_B_ratio3to1.csv",
    index=False
)


# ============================================================
# 9. ENSEMBLE ALL 20 MODELS
# ============================================================

all_model_columns = list(all_predictions.keys())

prediction_matrix["ENSEMBLE_MEAN_20"] = (
    prediction_matrix[all_model_columns].mean(axis=1)
)

submission_20 = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_MEAN_20"
    ].values

})

submission_20.to_csv(
    SUBMISSION_DIR / "submission_ensemble_20.csv",
    index=False
)


# ============================================================
# 10. MODEL DISAGREEMENT
# ============================================================

prediction_matrix["MODEL_PREDICTION_STD"] = (
    prediction_matrix[all_model_columns].std(
        axis=1,
        ddof=0
    )
)


# Save complete prediction matrix
prediction_matrix.to_csv(
    SUBMISSION_DIR / "all_20_model_predictions.csv",
    index=False
)


# ============================================================
# 11. SAVE PREDICTION SUMMARY
# ============================================================

summary_rows = []

for model_name, probabilities in all_predictions.items():

    summary_rows.append({

        "model_name": model_name,

        "group": model_groups[model_name],

        "prediction_mean": probabilities.mean(),

        "prediction_std": probabilities.std(),

        "prediction_min": probabilities.min(),

        "prediction_max": probabilities.max()

    })


summary_df = pd.DataFrame(summary_rows)

summary_df.to_csv(
    SUBMISSION_DIR / "test_prediction_summary.csv",
    index=False
)


# ============================================================
# 12. FINAL REPORT
# ============================================================

print("\n" + "=" * 75)
print("TEST SUBMISSION GENERATION COMPLETED SUCCESSFULLY")
print("=" * 75)

print(f"\nTest rows: {len(test_df):,}")

print(f"Individual model submissions: {len(all_predictions)}")

print("\nGroup A:")
print(f"Models: {len(group_A_models)}")
print(
    "Mean prediction:",
    f"{submission_A[TARGET_COL].mean():.6f}"
)

print("\nGroup B:")
print(f"Models: {len(group_B_models)}")
print(
    "Mean prediction:",
    f"{submission_B[TARGET_COL].mean():.6f}"
)

print("\nEnsemble of all 20 models:")
print(
    "Mean prediction:",
    f"{submission_20[TARGET_COL].mean():.6f}"
)

print(
    "Prediction disagreement mean:",
    f"{prediction_matrix['MODEL_PREDICTION_STD'].mean():.6f}"
)

print("\nOutput directory:")
print(SUBMISSION_DIR)

print("\nGenerated files:")

print("- 20 individual model submissions")

print("- submission_ensemble_A_original.csv")

print("- submission_ensemble_B_ratio3to1.csv")

print("- submission_ensemble_20.csv")

print("- all_20_model_predictions.csv")

print("- test_prediction_summary.csv")

print("\nAll files saved successfully.")
```
