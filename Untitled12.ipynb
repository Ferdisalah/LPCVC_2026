```python
# ============================================================
# HOME CREDIT
# 20 CATBOOST MODELS + 20 ROOT DECISION VECTORS
#
# GROUP A: 10 models, original class distribution
#          2 repetitions x 5-fold StratifiedKFold
#          each model trains on 80% of the data
#
# GROUP B: 10 models, negative:positive ratio = 3:1
#          disjoint positive samples and disjoint negative samples
#
# OUTPUT:
#   20 CatBoost models (.cbm)
#   20 JSON model exports
#   20 binary root vectors, each length 900
# ============================================================

import os
import sys
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold

try:
    from catboost import CatBoostClassifier, Pool
except ImportError:
    os.system(f"{sys.executable} -m pip -q install catboost")
    from catboost import CatBoostClassifier, Pool


# ============================================================
# 1. CONFIGURATION
# ============================================================

TRAIN_PATH = "/content/drive/MyDrive/Competition/input/train.csv"

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/Competition/output/"
    "CATBOOST_20_ROOT_VECTORS"
)

MODEL_DIR = OUTPUT_DIR / "models"
VECTOR_DIR = OUTPUT_DIR / "vectors"
INDIVIDUAL_DIR = VECTOR_DIR / "individual"

for folder in [OUTPUT_DIR, MODEL_DIR, VECTOR_DIR, INDIVIDUAL_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

TARGET = "TARGET"
ID_COL = "SK_ID_CURR"

SEED = 42

N_MODELS = 20
N_MODELS_A = 10
N_MODELS_B = 10

N_SPLITS = 5
N_REPEATS_A = 2

# Exactly 900 trees in each model.
CB_ITERATIONS = 900
CB_DEPTH = 7
CB_LEARNING_RATE = 0.04
CB_L2_LEAF_REG = 10.0
CB_VERBOSE = 200

CAT_FEATURES = [
    "NAME_CONTRACT_TYPE",
    "CODE_GENDER",
    "NAME_INCOME_TYPE",
    "NAME_EDUCATION_TYPE",
    "NAME_FAMILY_STATUS",
    "NAME_HOUSING_TYPE",
    "OCCUPATION_TYPE",
    "ORGANIZATION_TYPE",
]

# CatBoost JSON split types.
NUMERICAL_TYPES = {
    "FloatFeature"
}

CATEGORICAL_TYPES = {
    "OneHotFeature",
    "OnlineCtr",
}

random.seed(SEED)
np.random.seed(SEED)

warnings.filterwarnings("ignore", category=FutureWarning)


# ============================================================
# 2. LOAD DATA
# ============================================================

print("\n[1/6] Loading data...")

train = pd.read_csv(TRAIN_PATH)

required = [
    TARGET,
    ID_COL,
    "AMT_CREDIT",
    "AMT_ANNUITY",
] + CAT_FEATURES

missing = [c for c in required if c not in train.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

if train[ID_COL].duplicated().any():
    raise ValueError("Duplicate SK_ID_CURR values detected.")

y = train[TARGET].astype(np.int64).to_numpy()
train_ids = train[ID_COL].to_numpy()

if set(np.unique(y)) != {0, 1}:
    raise ValueError("TARGET must contain only 0 and 1.")

print("Train shape:", train.shape)
print("Non-default:", int((y == 0).sum()))
print("Default:", int((y == 1).sum()))
print("Default rate:", round(y.mean() * 100, 4), "%")


# ============================================================
# 3. PREPARE CATBOOST FEATURES
# ============================================================

print("\n[2/6] Preparing CatBoost features...")

# All original predictors except TARGET and ID.
cb_feature_cols = [
    c for c in train.columns
    if c not in [TARGET, ID_COL]
]

X_cb = train[cb_feature_cols].copy()

# Add CREDIT_TERM exactly as in the supplied script.
credit = pd.to_numeric(
    train["AMT_CREDIT"],
    errors="coerce"
)

annuity = pd.to_numeric(
    train["AMT_ANNUITY"],
    errors="coerce"
).replace(0, np.nan)

X_cb["CREDIT_TERM"] = credit / annuity

X_cb["CREDIT_TERM"] = X_cb["CREDIT_TERM"].replace(
    [np.inf, -np.inf],
    np.nan
)

# Handle categorical missing values.
for col in CAT_FEATURES:
    X_cb[col] = (
        X_cb[col]
        .where(X_cb[col].notna(), "__MISSING__")
        .astype(str)
    )

assert TARGET not in X_cb.columns
assert ID_COL not in X_cb.columns

print("CatBoost input shape:", X_cb.shape)
print("Categorical features:", len(CAT_FEATURES))
print("Added feature: CREDIT_TERM")


# ============================================================
# 4. CREATE 20 TRAINING SAMPLES
# ============================================================

print("\n[3/6] Creating model training samples...")

model_specs = []


# ------------------------------------------------------------
# GROUP A
# 10 models with original class distribution.
# 2 repetitions x 5 folds.
# Each model trains on 80% of the original dataset.
# ------------------------------------------------------------

for repeat in range(N_REPEATS_A):

    splitter = StratifiedKFold(
        n_splits=N_SPLITS,
        shuffle=True,
        random_state=SEED + repeat,
    )

    for fold, (fit_idx, valid_idx) in enumerate(
        splitter.split(X_cb, y)
    ):

        model_name = (
            f"A_original_r{repeat + 1}_f{fold + 1}"
        )

        model_specs.append({
            "model_name": model_name,
            "group": "A_original",
            "fit_idx": np.asarray(fit_idx, dtype=np.int64),
            "seed": SEED + 100 * repeat + fold,
            "sampling": "Original distribution, 80% fit split",
        })


# ------------------------------------------------------------
# GROUP B
# 10 models with:
# Non-default : Default = 3 : 1
#
# Positive rows are divided into 10 disjoint groups.
# Negative rows are also divided into disjoint groups.
#
# No training row is shared between two GROUP B models.
# ------------------------------------------------------------

positive_idx = np.flatnonzero(y == 1)
negative_idx = np.flatnonzero(y == 0)

rng = np.random.default_rng(SEED + 5000)

positive_idx = rng.permutation(positive_idx)
negative_idx = rng.permutation(negative_idx)

positive_chunks = np.array_split(
    positive_idx,
    N_MODELS_B
)

negative_cursor = 0

for model_no, positive_chunk in enumerate(
    positive_chunks,
    start=1
):

    n_positive = len(positive_chunk)
    n_negative = 3 * n_positive

    next_cursor = negative_cursor + n_negative

    if next_cursor > len(negative_idx):
        raise ValueError(
            "Not enough negative samples for disjoint 3:1 sampling."
        )

    negative_chunk = negative_idx[
        negative_cursor:next_cursor
    ]

    negative_cursor = next_cursor

    fit_idx = np.concatenate([
        positive_chunk,
        negative_chunk
    ])

    # Shuffle rows inside this model's training sample.
    model_rng = np.random.default_rng(
        SEED + 6000 + model_no
    )

    fit_idx = model_rng.permutation(fit_idx)

    model_specs.append({
        "model_name": f"B_ratio3to1_{model_no:02d}",
        "group": "B_ratio3to1",
        "fit_idx": fit_idx.astype(np.int64),
        "seed": SEED + 2000 + model_no,
        "sampling": "Disjoint samples with negative:positive = 3:1",
    })


assert len(model_specs) == 20

print("Total model specifications:", len(model_specs))


# ============================================================
# 5. VERIFY TRAINING SAMPLES AND SAVE MANIFEST
# ============================================================

print("\n[4/6] Verifying samples...")

manifest_rows = []
summary_rows = []

sample_sets = {}

for spec in model_specs:

    idx = spec["fit_idx"]
    subset_y = y[idx]

    n_positive = int((subset_y == 1).sum())
    n_negative = int((subset_y == 0).sum())

    if spec["group"] == "B_ratio3to1":

        assert n_negative == 3 * n_positive

    row_ids = train_ids[idx]

    sample_sets[spec["model_name"]] = set(
        row_ids.tolist()
    )

    summary_rows.append({
        "model_name": spec["model_name"],
        "group": spec["group"],
        "train_rows": len(idx),
        "non_default_rows": n_negative,
        "default_rows": n_positive,
        "negative_positive_ratio": n_negative / n_positive,
        "default_rate": n_positive / len(idx),
        "seed": spec["seed"],
    })

    for row_idx in idx:

        manifest_rows.append({
            "model_name": spec["model_name"],
            "group": spec["group"],
            ID_COL: train_ids[row_idx],
            TARGET: int(y[row_idx]),
        })


summary_df = pd.DataFrame(summary_rows)

summary_df.to_csv(
    OUTPUT_DIR / "model_training_summary.csv",
    index=False
)

manifest_df = pd.DataFrame(manifest_rows)

manifest_df.to_csv(
    OUTPUT_DIR / "training_sample_manifest.csv",
    index=False
)

print(summary_df.to_string(index=False))


# ------------------------------------------------------------
# Verify zero overlap inside GROUP B.
# ------------------------------------------------------------

group_b_names = [
    s["model_name"]
    for s in model_specs
    if s["group"] == "B_ratio3to1"
]

for i in range(len(group_b_names)):

    for j in range(i + 1, len(group_b_names)):

        set_i = sample_sets[group_b_names[i]]
        set_j = sample_sets[group_b_names[j]]

        overlap = set_i.intersection(set_j)

        if len(overlap) > 0:
            raise RuntimeError(
                f"Overlap found between "
                f"{group_b_names[i]} and {group_b_names[j]}"
            )

print("\nGROUP B: zero training-row overlap verified.")
```

```python

# ============================================================
# 6. PAIRWISE TRAINING OVERLAP REPORT
# ============================================================

print("\nCreating pairwise overlap report...")

overlap_rows = []

for i in range(len(model_specs)):

    name_i = model_specs[i]["model_name"]
    set_i = sample_sets[name_i]

    for j in range(i + 1, len(model_specs)):

        name_j = model_specs[j]["model_name"]
        set_j = sample_sets[name_j]

        intersection = len(set_i.intersection(set_j))
        union = len(set_i.union(set_j))

        overlap_rows.append({
            "model_1": name_i,
            "model_2": name_j,
            "group_1": model_specs[i]["group"],
            "group_2": model_specs[j]["group"],
            "overlap_rows": intersection,
            "overlap_fraction_smaller": (
                intersection / min(len(set_i), len(set_j))
            ),
            "jaccard_overlap": intersection / union if union > 0 else 0.0,
        })

pd.DataFrame(overlap_rows).to_csv(
    OUTPUT_DIR / "pairwise_train_overlap.csv",
    index=False
)


# ============================================================
# 7. ROOT SPLIT EXTRACTION
# ============================================================

def extract_root_vector(json_path):

    with open(json_path, "r", encoding="utf-8") as f:
        model_json = json.load(f)

    trees = model_json.get("oblivious_trees")

    if trees is None:
        raise RuntimeError(
            f"Missing oblivious_trees in {json_path}"
        )

    if len(trees) != CB_ITERATIONS:
        raise RuntimeError(
            f"Expected {CB_ITERATIONS} trees, "
            f"found {len(trees)}"
        )

    # Encoding:
    # 0 = Numerical root split
    # 1 = Categorical root split
    # 2 = No split

    vector = np.full(
        CB_ITERATIONS,
        2,
        dtype=np.uint8
    )

    audit = []

    for tree_idx, tree in enumerate(trees):

        splits = tree.get("splits") or []

        # ----------------------------------------------------
        # CASE 1: TREE WITHOUT SPLIT
        # ----------------------------------------------------

        if len(splits) == 0:

            vector[tree_idx] = 2

            audit.append({
                "tree_index": tree_idx + 1,
                "root_value": 2,
                "root_kind": "no_split",
                "root_split_type": None,
                "root_split_details": None,
                "has_root_split": False,
            })

            continue

        # ----------------------------------------------------
        # CASE 2: TREE HAS SPLITS
        # ----------------------------------------------------

        # CatBoost symmetric-tree JSON stores split levels
        # in reverse order. The LAST element is the root.

        root_split = splits[-1]

        split_type = root_split.get("split_type")

        if split_type in NUMERICAL_TYPES:

            value = 0
            kind = "numerical"

        elif split_type in CATEGORICAL_TYPES:

            value = 1
            kind = "categorical"

        else:

            raise RuntimeError(
                f"Unknown root split type: {split_type} "
                f"in tree {tree_idx + 1}, file: {json_path}"
            )

        vector[tree_idx] = value

        audit.append({
            "tree_index": tree_idx + 1,
            "root_value": value,
            "root_kind": kind,
            "root_split_type": split_type,
            "root_split_details": json.dumps(
                root_split,
                ensure_ascii=False
            ),
            "has_root_split": True,
        })

    # --------------------------------------------------------
    # FINAL VECTOR VALIDATION
    # --------------------------------------------------------

    assert vector.shape == (CB_ITERATIONS,)

    assert np.isin(vector, [0, 1, 2]).all()

    assert len(audit) == CB_ITERATIONS

    return vector, audit


# ============================================================
# 8. TRAIN THE 20 MODELS
# ============================================================

print("\n" + "=" * 75)
print("TRAINING 20 CATBOOST MODELS")
print("=" * 75)

all_vectors = []
all_audits = []
vector_summary = []

for model_number, spec in enumerate(
    model_specs,
    start=1
):

    model_name = spec["model_name"]
    fit_idx = spec["fit_idx"]

    print("\n" + "-" * 75)
    print(f"MODEL {model_number}/20: {model_name}")
    print("Group:", spec["group"])
    print("Training rows:", len(fit_idx))
    print("Non-default:", int((y[fit_idx] == 0).sum()))
    print("Default:", int((y[fit_idx] == 1).sum()))

    train_pool = Pool(
        X_cb.iloc[fit_idx],
        y[fit_idx],
        cat_features=CAT_FEATURES,
    )

    cb_model = CatBoostClassifier(
        iterations=CB_ITERATIONS,
        depth=CB_DEPTH,
        learning_rate=CB_LEARNING_RATE,
        l2_leaf_reg=CB_L2_LEAF_REG,
        loss_function="Logloss",
        eval_metric="AUC",
        random_seed=int(spec["seed"]),
        verbose=CB_VERBOSE,
        allow_writing_files=False,
        thread_count=-1,
        grow_policy="SymmetricTree",
    )

    # No validation-based early stopping.
    # Every model must contain exactly CB_ITERATIONS trees.

    cb_model.fit(train_pool)

    if cb_model.tree_count_ != CB_ITERATIONS:

        raise RuntimeError(
            f"{model_name}: expected {CB_ITERATIONS} trees, "
            f"got {cb_model.tree_count_}"
        )

    # --------------------------------------------------------
    # Save native CatBoost model.
    # --------------------------------------------------------

    model_path = MODEL_DIR / f"{model_name}.cbm"

    cb_model.save_model(
        str(model_path),
        format="cbm"
    )

    # --------------------------------------------------------
    # Export JSON for root split extraction.
    # --------------------------------------------------------

    json_path = MODEL_DIR / f"{model_name}.json"

    cb_model.save_model(
        str(json_path),
        format="json"
    )

    # --------------------------------------------------------
    # Extract root vectors.
    # 0 = Numerical
    # 1 = Categorical
    # 2 = No split
    # --------------------------------------------------------

    vector, audit = extract_root_vector(json_path)

    for row in audit:

        row["model_name"] = model_name
        row["group"] = spec["group"]

    all_vectors.append(vector)
    all_audits.extend(audit)

    # --------------------------------------------------------
    # Save one vector per model.
    # --------------------------------------------------------

    np.save(
        INDIVIDUAL_DIR / f"{model_name}_vector.npy",
        vector
    )

    vector_columns = [
        f"TREE_{i:03d}"
        for i in range(1, CB_ITERATIONS + 1)
    ]

    pd.DataFrame(
        [vector],
        columns=vector_columns
    ).to_csv(
        INDIVIDUAL_DIR / f"{model_name}_vector.csv",
        index=False
    )

    # --------------------------------------------------------
    # Count the three root types independently.
    # --------------------------------------------------------

    n_num = int(np.sum(vector == 0))
    n_cat = int(np.sum(vector == 1))
    n_no_split = int(np.sum(vector == 2))

    vector_summary.append({

        "model_name": model_name,
        "group": spec["group"],
        "train_rows": len(fit_idx),

        "numerical_root_count": n_num,
        "categorical_root_count": n_cat,
        "no_split_tree_count": n_no_split,

        "numerical_fraction": n_num / CB_ITERATIONS,
        "categorical_fraction": n_cat / CB_ITERATIONS,
        "no_split_fraction": n_no_split / CB_ITERATIONS,

    })

    print("Model saved:", model_path)
    print("Vector shape:", vector.shape)

    print("Numerical roots:", n_num)
    print("Categorical roots:", n_cat)
    print("Trees without splits:", n_no_split)

    print(
        "Total:",
        n_num + n_cat + n_no_split
    )

    del train_pool
    del cb_model


# ============================================================
# 9. SAVE THE 20 VECTORS
# ============================================================

print("\nSaving the 20 root vectors...")

root_vectors = np.stack(
    all_vectors,
    axis=0
).astype(np.uint8)

if root_vectors.shape != (20, CB_ITERATIONS):

    raise RuntimeError(
        f"Expected (20, {CB_ITERATIONS}), "
        f"got {root_vectors.shape}"
    )

# Validate all three encodings.

if not np.isin(root_vectors, [0, 1, 2]).all():

    raise RuntimeError(
        "Invalid values found in root_vectors."
    )

# Main matrix: 20 models x 900 trees.

np.save(
    VECTOR_DIR / "root_vectors.npy",
    root_vectors
)

# Readable CSV with one row per model.

vector_columns = [
    f"TREE_{i:03d}"
    for i in range(1, CB_ITERATIONS + 1)
]

root_df = pd.DataFrame(
    root_vectors,
    columns=vector_columns
)

root_df.insert(
    0,
    "model_name",
    [s["model_name"] for s in model_specs]
)

root_df.insert(
    1,
    "group",
    [s["group"] for s in model_specs]
)

root_df.to_csv(
    VECTOR_DIR / "root_vectors.csv",
    index=False
)


# ============================================================
# 10. SAVE TREE-LEVEL AUDIT
# ============================================================

audit_df = pd.DataFrame(all_audits)

audit_df.to_csv(
    OUTPUT_DIR / "root_split_audit.csv",
    index=False
)


# ============================================================
# 11. SAVE VECTOR SUMMARY
# ============================================================

summary_vectors_df = pd.DataFrame(vector_summary)

summary_vectors_df.to_csv(
    VECTOR_DIR / "root_vector_summary.csv",
    index=False
)

print("\nRoot vector summary:")

print(
    summary_vectors_df[
        [
            "model_name",
            "group",
            "numerical_root_count",
            "categorical_root_count",
            "no_split_tree_count",
        ]
    ].to_string(index=False)
)


# ============================================================
# 12. SAVE MODEL ORDER
# ============================================================

order_df = pd.DataFrame({

    "matrix_row": np.arange(1, 21),

    "model_name": [
        s["model_name"] for s in model_specs
    ],

    "group": [
        s["group"] for s in model_specs
    ],

})

order_df.to_csv(
    VECTOR_DIR / "root_vector_model_order.csv",
    index=False
)


# ============================================================
# 13. SAVE EXPERIMENT METADATA
# ============================================================

metadata = {

    "total_models": 20,
    "group_a_models": 10,
    "group_b_models": 10,

    "trees_per_model": CB_ITERATIONS,

    "root_vector_shape": [
        20,
        CB_ITERATIONS
    ],

    "encoding": {

        "0": "Numerical root split",

        "1": "Categorical root split",

        "2": "Tree without split",

    },

    "categorical_split_types": sorted(
        CATEGORICAL_TYPES
    ),

    "numerical_split_types": sorted(
        NUMERICAL_TYPES
    ),

    "catboost_parameters": {

        "iterations": CB_ITERATIONS,

        "depth": CB_DEPTH,

        "learning_rate": CB_LEARNING_RATE,

        "l2_leaf_reg": CB_L2_LEAF_REG,

        "loss_function": "Logloss",

        "eval_metric": "AUC",

        "grow_policy": "SymmetricTree",

        "class_weights": None,

    },

    "group_a_sampling": (

        "Two repetitions of 5-fold StratifiedKFold; "
        "10 models, each trained on 80%."

    ),

    "group_b_sampling": (

        "10 models with negative:positive ratio 3:1; "
        "positive and negative training rows are disjoint "
        "between group B models."

    ),

    "training_rows_group_b_unused_negatives": int(
        len(negative_idx) - negative_cursor
    ),

    "cross_group_overlap_allowed": True,

    "seed": SEED,

    "model_order": [
        s["model_name"] for s in model_specs
    ],

}

with open(
    OUTPUT_DIR / "experiment_metadata.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2,
        ensure_ascii=False
    )


# ============================================================
# 14. FINAL VALIDATION
# ============================================================

print("\nFinal validation...")

assert len(list(MODEL_DIR.glob("*.cbm"))) == 20

assert len(list(MODEL_DIR.glob("*.json"))) == 20

assert len(
    list(INDIVIDUAL_DIR.glob("*_vector.npy"))
) == 20

assert root_vectors.shape == (20, CB_ITERATIONS)

assert np.isin(root_vectors, [0, 1, 2]).all()

assert len(all_audits) == 20 * CB_ITERATIONS

# Verify that every model has exactly CB_ITERATIONS entries.

for i in range(20):

    assert len(root_vectors[i]) == CB_ITERATIONS

    assert (
        np.sum(root_vectors[i] == 0)
        + np.sum(root_vectors[i] == 1)
        + np.sum(root_vectors[i] == 2)
    ) == CB_ITERATIONS


print("\n" + "=" * 75)
print("EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 75)

print("CatBoost models:", len(list(MODEL_DIR.glob("*.cbm"))))

print("JSON models:", len(list(MODEL_DIR.glob("*.json"))))

print(
    "Individual vectors:",
    len(list(INDIVIDUAL_DIR.glob("*_vector.npy")))
)

print("Combined vector shape:", root_vectors.shape)

print("\nEncoding:")
print("0 = Numerical")
print("1 = Categorical")
print("2 = No split")

print("\nTotal numerical roots:",
      int(np.sum(root_vectors == 0)))

print("Total categorical roots:",
      int(np.sum(root_vectors == 1)))

print("Total trees without splits:",
      int(np.sum(root_vectors == 2)))

print("\nMain output directory:")
print(OUTPUT_DIR)

print("\nVector matrix:")
print(VECTOR_DIR / "root_vectors.npy")

print("\nReadable vectors:")
print(VECTOR_DIR / "root_vectors.csv")

print("\nRoot split audit:")
print(OUTPUT_DIR / "root_split_audit.csv")

print("\nTraining sample manifest:")
print(OUTPUT_DIR / "training_sample_manifest.csv")

print("\nPairwise overlap:")
print(OUTPUT_DIR / "pairwise_train_overlap.csv")
```

```python

# ============================================================
# 15. TEST PREDICTIONS + 20 INDIVIDUAL SUBMISSIONS + ENSEMBLES
# HOME CREDIT - 20 CATBOOST MODELS
# ============================================================

import re
import gc
import numpy as np
import pandas as pd

from pathlib import Path
from catboost import CatBoostClassifier


print("\n" + "=" * 75)
print("GENERATING TEST SUBMISSIONS FOR ALL 20 CATBOOST MODELS")
print("=" * 75)


# ============================================================
# 1. CONFIGURATION
# ============================================================

MODEL_ROOT = Path(
    "/content/drive/MyDrive/Competition/output/"
    "CATBOOST_20_ROOT_VECTORS"
)

MODELS_DIR = MODEL_ROOT / "models"

SUBMISSION_DIR = MODEL_ROOT / "test_submissions"

SUBMISSION_DIR.mkdir(
    parents=True,
    exist_ok=True
)

ID_COL = "SK_ID_CURR"
TARGET_COL = "TARGET"

EXPECTED_MODELS = 20


# ============================================================
# 2. GET TEST DATA
# ============================================================

# Your original test dataset is named "test".

if "test" in globals():

    test_df = test.copy()

elif "test_df" in globals():

    test_df = test_df.copy()

else:

    raise NameError(
        "Neither 'test' nor 'test_df' exists. "
        "Load competition test.csv first."
    )


# Validate ID

if ID_COL not in test_df.columns:

    raise ValueError(
        f"Test dataset is missing {ID_COL}."
    )


# Validate TARGET absence

if TARGET_COL in test_df.columns:

    raise ValueError(
        "TARGET must not exist in the test dataset."
    )


# ============================================================
# 3. CREATE CREDIT_TERM
# ============================================================

# IMPORTANT:
# This uses the standard training definition:
#
# CREDIT_TERM = AMT_CREDIT / AMT_ANNUITY
#
# It must match the exact definition used during training.

if "CREDIT_TERM" not in test_df.columns:

    required_columns = [
        "AMT_CREDIT",
        "AMT_ANNUITY"
    ]

    missing_columns = [
        col for col in required_columns
        if col not in test_df.columns
    ]

    if missing_columns:

        raise ValueError(
            "Cannot create CREDIT_TERM. "
            f"Missing columns: {missing_columns}"
        )

    credit = pd.to_numeric(
        test_df["AMT_CREDIT"],
        errors="coerce"
    )

    annuity = pd.to_numeric(
        test_df["AMT_ANNUITY"],
        errors="coerce"
    )

    # Avoid division by zero
    annuity = annuity.where(
        annuity != 0,
        np.nan
    )

    test_df["CREDIT_TERM"] = credit / annuity

    # Replace infinite values with NaN
    test_df["CREDIT_TERM"] = (
        test_df["CREDIT_TERM"]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
    )

    print("\nCREDIT_TERM successfully created.")

else:

    print("\nCREDIT_TERM already exists in test data.")


print(
    "CREDIT_TERM missing values:",
    test_df["CREDIT_TERM"].isna().sum()
)

print(
    "CREDIT_TERM mean:",
    test_df["CREDIT_TERM"].mean()
)

print(f"\nTest rows: {len(test_df):,}")
print(f"Test columns: {len(test_df.columns)}")


# ============================================================
# 4. VALIDATE MODEL DIRECTORY
# ============================================================

if not MODELS_DIR.exists():

    raise FileNotFoundError(
        f"Models directory does not exist:\n{MODELS_DIR}"
    )


model_paths = list(
    MODELS_DIR.glob("*.cbm")
)


def natural_sort_key(path):

    return [
        int(part) if part.isdigit() else part.lower()
        for part in re.split(r"(\d+)", path.stem)
    ]


model_paths = sorted(
    model_paths,
    key=natural_sort_key
)


if len(model_paths) != EXPECTED_MODELS:

    raise RuntimeError(
        f"Expected {EXPECTED_MODELS} models, "
        f"but found {len(model_paths)}.\n"
        f"Directory: {MODELS_DIR}"
    )


print(f"\nFound {len(model_paths)} CatBoost models.")


# ============================================================
# 5. PRE-VALIDATE ALL MODELS AND FEATURES
# ============================================================

print("\nPre-validating all model features...")


model_feature_specs = {}


for model_path in model_paths:

    model_name = model_path.stem

    model = CatBoostClassifier()

    model.load_model(
        str(model_path)
    )

    feature_names = model.feature_names_

    if not feature_names:

        raise RuntimeError(
            f"No stored feature names in {model_name}."
        )

    missing_features = [
        feature
        for feature in feature_names
        if feature not in test_df.columns
    ]

    if missing_features:

        raise ValueError(
            f"\nMissing features for {model_name}:\n"
            f"{missing_features}\n\n"
            "Check test feature engineering."
        )

    model_feature_specs[model_name] = feature_names

    del model


gc.collect()


print(
    "All 20 models passed feature validation."
)


# ============================================================
# 6. PREDICT WITH EACH MODEL
# ============================================================

all_predictions = {}

model_groups = {}


for model_idx, model_path in enumerate(
    model_paths,
    start=1
):

    model_name = model_path.stem

    print("\n" + "-" * 65)

    print(
        f"MODEL {model_idx}/{EXPECTED_MODELS}: "
        f"{model_name}"
    )


    # --------------------------------------------------------
    # Load model
    # --------------------------------------------------------

    model = CatBoostClassifier()

    model.load_model(
        str(model_path)
    )


    # --------------------------------------------------------
    # Retrieve exact training feature order
    # --------------------------------------------------------

    feature_names = model_feature_specs[model_name]

    X_test = test_df[feature_names]


    # --------------------------------------------------------
    # Identify TARGET = 1 class
    # --------------------------------------------------------

    classes = list(model.classes_)

    positive_class_indices = [
        i
        for i, cls in enumerate(classes)
        if cls == 1 or str(cls) == "1"
    ]

    if len(positive_class_indices) != 1:

        raise RuntimeError(
            f"Could not identify TARGET=1 for "
            f"{model_name}. Classes: {classes}"
        )

    positive_class_idx = positive_class_indices[0]


    # --------------------------------------------------------
    # Predict probabilities
    # --------------------------------------------------------

    probabilities = model.predict_proba(
        X_test
    )[:, positive_class_idx]

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )


    # --------------------------------------------------------
    # Validate predictions
    # --------------------------------------------------------

    if len(probabilities) != len(test_df):

        raise RuntimeError(
            f"Prediction length mismatch: {model_name}"
        )

    if not np.isfinite(probabilities).all():

        raise RuntimeError(
            f"Invalid predictions detected: {model_name}"
        )

    if (
        probabilities.min() < 0
        or probabilities.max() > 1
    ):

        raise RuntimeError(
            f"Probabilities outside [0, 1]: {model_name}"
        )


    # --------------------------------------------------------
    # Store probabilities
    # --------------------------------------------------------

    all_predictions[model_name] = probabilities


    # --------------------------------------------------------
    # Identify model group
    # --------------------------------------------------------

    if model_name.startswith("A_original_"):

        model_groups[model_name] = "A_original"

    elif model_name.startswith("B_ratio3to1_"):

        model_groups[model_name] = "B_ratio3to1"

    else:

        raise ValueError(
            f"Unknown model group: {model_name}"
        )


    # --------------------------------------------------------
    # Save individual submission
    # --------------------------------------------------------

    individual_submission = pd.DataFrame({

        ID_COL: test_df[ID_COL].values,

        TARGET_COL: probabilities

    })


    individual_path = (
        SUBMISSION_DIR /
        f"submission_{model_name}.csv"
    )


    individual_submission.to_csv(
        individual_path,
        index=False
    )


    print(
        f"Saved: {individual_path.name}"
    )

    print(
        f"Mean: {probabilities.mean():.6f} | "
        f"Std: {probabilities.std():.6f} | "
        f"Min: {probabilities.min():.6f} | "
        f"Max: {probabilities.max():.6f}"
    )


    del model
    del X_test
    del individual_submission

    gc.collect()


# ============================================================
# 7. VALIDATE MODEL GROUPS
# ============================================================

group_A_models = [
    name
    for name, group in model_groups.items()
    if group == "A_original"
]

group_B_models = [
    name
    for name, group in model_groups.items()
    if group == "B_ratio3to1"
]


if len(group_A_models) != 10:

    raise RuntimeError(
        f"Expected 10 A_original models, "
        f"found {len(group_A_models)}."
    )


if len(group_B_models) != 10:

    raise RuntimeError(
        f"Expected 10 B_ratio3to1 models, "
        f"found {len(group_B_models)}."
    )


# ============================================================
# 8. BUILD COMPLETE PREDICTION MATRIX
# ============================================================

prediction_matrix = pd.DataFrame({

    ID_COL: test_df[ID_COL].values

})


for model_name, probabilities in all_predictions.items():

    prediction_matrix[model_name] = probabilities


# ============================================================
# 9. ENSEMBLE A: ORIGINAL DISTRIBUTION
# ============================================================

prediction_matrix["ENSEMBLE_A_ORIGINAL"] = (
    prediction_matrix[group_A_models].mean(axis=1)
)


submission_A = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_A_ORIGINAL"
    ].values

})


submission_A.to_csv(
    SUBMISSION_DIR / "submission_ensemble_A_original.csv",
    index=False
)


# ============================================================
# 10. ENSEMBLE B: RATIO 3:1
# ============================================================

prediction_matrix["ENSEMBLE_B_RATIO3TO1"] = (
    prediction_matrix[group_B_models].mean(axis=1)
)


submission_B = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_B_RATIO3TO1"
    ].values

})


submission_B.to_csv(
    SUBMISSION_DIR / "submission_ensemble_B_ratio3to1.csv",
    index=False
)


# ============================================================
# 11. ENSEMBLE ALL 20 MODELS
# ============================================================

all_model_columns = list(all_predictions.keys())


prediction_matrix["ENSEMBLE_MEAN_20"] = (
    prediction_matrix[all_model_columns].mean(axis=1)
)


submission_20 = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_MEAN_20"
    ].values

})


submission_20.to_csv(
    SUBMISSION_DIR / "submission_ensemble_20.csv",
    index=False
)


# ============================================================
# 12. MODEL DISAGREEMENT
# ============================================================

prediction_matrix["MODEL_PREDICTION_STD"] = (
    prediction_matrix[all_model_columns].std(
        axis=1,
        ddof=0
    )
)


# Save all predictions
prediction_matrix.to_csv(
    SUBMISSION_DIR / "all_20_model_predictions.csv",
    index=False
)


# ============================================================
# 13. SAVE TEST PREDICTION SUMMARY
# ============================================================

summary_rows = []


for model_name, probabilities in all_predictions.items():

    summary_rows.append({

        "model_name": model_name,

        "group": model_groups[model_name],

        "prediction_mean": probabilities.mean(),

        "prediction_std": probabilities.std(),

        "prediction_min": probabilities.min(),

        "prediction_max": probabilities.max()

    })


summary_df = pd.DataFrame(summary_rows)


summary_df.to_csv(
    SUBMISSION_DIR / "test_prediction_summary.csv",
    index=False
)


# ============================================================
# 14. FINAL VALIDATION
# ============================================================

submission_files = [
    SUBMISSION_DIR / f"submission_{name}.csv"
    for name in all_predictions
]

submission_files.extend([

    SUBMISSION_DIR / "submission_ensemble_A_original.csv",

    SUBMISSION_DIR / "submission_ensemble_B_ratio3to1.csv",

    SUBMISSION_DIR / "submission_ensemble_20.csv"

])


for file_path in submission_files:

    if not file_path.exists():

        raise RuntimeError(
            f"Expected output file not found: {file_path}"
        )

    check = pd.read_csv(file_path)

    if list(check.columns) != [ID_COL, TARGET_COL]:

        raise RuntimeError(
            f"Invalid submission columns: {file_path.name}"
        )

    if len(check) != len(test_df):

        raise RuntimeError(
            f"Invalid submission row count: {file_path.name}"
        )

    if not np.isfinite(
        check[TARGET_COL].to_numpy()
    ).all():

        raise RuntimeError(
            f"Invalid TARGET probabilities: {file_path.name}"
        )


# ============================================================
# 15. FINAL REPORT
# ============================================================

print("\n" + "=" * 75)

print("TEST SUBMISSION GENERATION COMPLETED SUCCESSFULLY")

print("=" * 75)


print(f"\nTest rows: {len(test_df):,}")

print(
    f"Individual model submissions: "
    f"{len(all_predictions)}"
)


print("\nGROUP A: ORIGINAL DISTRIBUTION")

print(f"Models: {len(group_A_models)}")

print(
    "Mean prediction:",
    f"{submission_A[TARGET_COL].mean():.6f}"
)


print("\nGROUP B: RATIO 3:1")

print(f"Models: {len(group_B_models)}")

print(
    "Mean prediction:",
    f"{submission_B[TARGET_COL].mean():.6f}"
)


print("\nENSEMBLE OF ALL 20 MODELS")

print(
    "Mean prediction:",
    f"{submission_20[TARGET_COL].mean():.6f}"
)

print(
    "Prediction disagreement mean:",
    f"{prediction_matrix['MODEL_PREDICTION_STD'].mean():.6f}"
)


print("\nOUTPUT DIRECTORY:")

print(SUBMISSION_DIR)


print("\nGENERATED FILES:")

print("- 20 individual model submissions")

print("- submission_ensemble_A_original.csv")

print("- submission_ensemble_B_ratio3to1.csv")

print("- submission_ensemble_20.csv")

print("- all_20_model_predictions.csv")

print("- test_prediction_summary.csv")


print("\nAll files saved and validated successfully.")
```

```python

# ============================================================
# EXPERIMENT 2: 20 CATBOOST MODELS
# A: ORIGINAL DISTRIBUTION (10 MODELS)
# B: FULL DEFAULT RETENTION + 3:1 (10 MODELS)
#
# ROOT VECTOR:
# 0 = Numerical root
# 1 = Categorical root
# 2 = No split
# ============================================================

import os
import re
import gc
import json
import numpy as np
import pandas as pd

from pathlib import Path
from catboost import CatBoostClassifier, Pool


# ============================================================
# 1. CHECK REQUIRED VARIABLES
# ============================================================

required_variables = [
    "X_cb",
    "y",
    "CAT_FEATURES",
    "CB_ITERATIONS",
    "CB_DEPTH",
    "CB_LEARNING_RATE",
    "CB_L2_LEAF_REG",
    "model_specs"
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        f"Missing required variables: {missing_variables}\n"
        "Run the data preparation and configuration sections first."
    )


# ============================================================
# 2. CONFIGURATION AND OUTPUT DIRECTORIES
# ============================================================

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/Competition/output/"
    "CATBOOST_20_ROOT_VECTORS_FULLPOS_3TO1"
)

MODEL_DIR = OUTPUT_DIR / "models"
VECTOR_DIR = OUTPUT_DIR / "vectors"
INDIVIDUAL_DIR = VECTOR_DIR / "individual"

for directory in [
    OUTPUT_DIR,
    MODEL_DIR,
    VECTOR_DIR,
    INDIVIDUAL_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

SEED_VALUE = int(globals().get("SEED", 42))
CB_VERBOSE_VALUE = globals().get("CB_VERBOSE", 200)

B_NEGATIVE_TO_POSITIVE_RATIO = 3

EXPECTED_MODELS = 20

NUMERICAL_TYPES = set(
    globals().get("NUMERICAL_TYPES", {"FloatFeature"})
)

CATEGORICAL_TYPES = set(
    globals().get(
        "CATEGORICAL_TYPES",
        {"OneHotFeature", "OnlineCtr"}
    )
)

if not NUMERICAL_TYPES:
    raise ValueError("NUMERICAL_TYPES is empty.")

if not CATEGORICAL_TYPES:
    raise ValueError("CATEGORICAL_TYPES is empty.")


# ============================================================
# 3. VALIDATE TRAINING DATA
# ============================================================

y_array = np.asarray(y).reshape(-1)

if len(X_cb) != len(y_array):
    raise ValueError(
        "X_cb and y have different lengths."
    )

if not np.isin(y_array, [0, 1]).all():
    raise ValueError(
        "Target must contain only 0 and 1."
    )

if not X_cb.columns.is_unique:
    raise ValueError(
        "Duplicate feature names detected in X_cb."
    )

if isinstance(CAT_FEATURES, (str, int)):
    CAT_FEATURES = [CAT_FEATURES]

missing_cat_features = [
    col for col in CAT_FEATURES
    if isinstance(col, str) and col not in X_cb.columns
]

if missing_cat_features:
    raise ValueError(
        f"Categorical features missing from X_cb: "
        f"{missing_cat_features}"
    )

print("=" * 75)
print("EXPERIMENT 2: DATA VALIDATION")
print("=" * 75)

print("Training rows:", len(X_cb))
print("Features:", X_cb.shape[1])
print("Non-default:", int(np.sum(y_array == 0)))
print("Default:", int(np.sum(y_array == 1)))
print("Iterations:", CB_ITERATIONS)


# ============================================================
# 4. PRESERVE ORIGINAL GROUP A MODELS
# ============================================================

group_A_specs = [
    dict(spec)
    for spec in model_specs
    if spec["group"] == "A_original"
]

if len(group_A_specs) != 10:
    raise RuntimeError(
        f"Expected 10 original A models, "
        f"found {len(group_A_specs)}."
    )

group_A_specs = sorted(
    group_A_specs,
    key=lambda s: (
        int(re.search(r"_r(\d+)_f(\d+)$", s["model_name"]).group(1)),
        int(re.search(r"_r(\d+)_f(\d+)$", s["model_name"]).group(2))
    )
)

for spec in group_A_specs:

    fit_idx = np.asarray(
        spec["fit_idx"],
        dtype=np.int64
    )

    if len(fit_idx) == 0:
        raise RuntimeError(
            f"Empty training indices: {spec['model_name']}"
        )

    if fit_idx.min() < 0 or fit_idx.max() >= len(y_array):
        raise IndexError(
            f"Invalid training index in {spec['model_name']}"
        )

    if len(np.unique(fit_idx)) != len(fit_idx):
        raise RuntimeError(
            f"Duplicate indices in {spec['model_name']}"
        )

    spec["fit_idx"] = fit_idx


# ============================================================
# 5. REBUILD GROUP B
# ALL DEFAULTS + 3 TIMES NON-DEFAULTS
# ============================================================

print("\n" + "=" * 75)
print("BUILDING GROUP B: FULL DEFAULT RETENTION")
print("=" * 75)

group_B_specs = []

for b_idx, a_spec in enumerate(
    group_A_specs,
    start=1
):

    a_name = a_spec["model_name"]

    a_fit_idx = a_spec["fit_idx"]

    a_targets = y_array[a_fit_idx]

    # Retain ALL defaults in the corresponding A training split.
    positive_idx = a_fit_idx[
        a_targets == 1
    ]

    # Available non-defaults in the same A training split.
    negative_idx = a_fit_idx[
        a_targets == 0
    ]

    n_positive = len(positive_idx)

    n_negative_required = (
        B_NEGATIVE_TO_POSITIVE_RATIO * n_positive
    )

    if n_positive == 0:
        raise RuntimeError(
            f"No default samples in {a_name}."
        )

    if n_negative_required > len(negative_idx):
        raise RuntimeError(
            f"Insufficient non-default samples for B model {b_idx}."
        )

    # Different negative samples for each B model.
    # No replacement inside an individual B model.
    # Reuse between different B models is allowed.
    sampling_rng = np.random.default_rng(
        SEED_VALUE + 10000 + b_idx
    )

    selected_negative_idx = sampling_rng.choice(
        negative_idx,
        size=n_negative_required,
        replace=False
    )

    # All defaults + selected non-defaults.
    b_fit_idx = np.concatenate([
        positive_idx,
        selected_negative_idx
    ])

    sampling_rng.shuffle(b_fit_idx)

    b_targets = y_array[b_fit_idx]

    b_positive_count = int(
        np.sum(b_targets == 1)
    )

    b_negative_count = int(
        np.sum(b_targets == 0)
    )

    # Validate counts and uniqueness.
    assert b_positive_count == n_positive

    assert (
        b_negative_count == n_negative_required
    )

    assert len(np.unique(b_fit_idx)) == len(b_fit_idx)

    assert (
        b_negative_count / b_positive_count
        == B_NEGATIVE_TO_POSITIVE_RATIO
    )

    b_model_name = f"B_ratio3to1_{b_idx:02d}"

    b_spec = {

        "model_name": b_model_name,

        "group": "B_ratio3to1",

        "fit_idx": b_fit_idx,

        "seed": SEED_VALUE + 20000 + b_idx,

        "paired_A_model": a_name,

        "positive_count": b_positive_count,

        "negative_count": b_negative_count

    }

    group_B_specs.append(b_spec)

    print(f"\n{b_model_name}")
    print("Paired A:", a_name)
    print("Defaults:", b_positive_count)
    print("Non-defaults:", b_negative_count)
    print("Total:", len(b_fit_idx))


# Final model order: A1...A10, B1...B10.
model_specs = group_A_specs + group_B_specs

if len(model_specs) != EXPECTED_MODELS:
    raise RuntimeError(
        f"Expected 20 models, got {len(model_specs)}."
    )


# ============================================================
# 6. BUILD TRAINING SAMPLE MANIFEST
# ============================================================

print("\nCreating training sample manifest...")

manifest_parts = []

sampling_summary_rows = []

sample_sets = {}

for spec in model_specs:

    model_name = spec["model_name"]
    fit_idx = spec["fit_idx"]

    targets = y_array[fit_idx]

    n_positive = int(np.sum(targets == 1))
    n_negative = int(np.sum(targets == 0))

    sample_sets[model_name] = set(
        fit_idx.tolist()
    )

    manifest_parts.append(
        pd.DataFrame({

            "model_name": model_name,

            "group": spec["group"],

            "row_index": fit_idx,

            "TARGET": targets,

            "paired_A_model": spec.get(
                "paired_A_model",
                None
            )

        })
    )

    sampling_summary_rows.append({

        "model_name": model_name,

        "group": spec["group"],

        "paired_A_model": spec.get(
            "paired_A_model",
            None
        ),

        "training_rows": len(fit_idx),

        "non_default": n_negative,

        "default": n_positive,

        "negative_to_positive_ratio": (
            n_negative / n_positive
        )

    })


manifest_df = pd.concat(
    manifest_parts,
    ignore_index=True
)

manifest_df.to_csv(
    OUTPUT_DIR / "training_sample_manifest.csv",
    index=False
)

sampling_summary_df = pd.DataFrame(
    sampling_summary_rows
)

sampling_summary_df.to_csv(
    OUTPUT_DIR / "training_sample_summary.csv",
    index=False
)

print("Training sample manifest saved.")


# ============================================================
# 7. PAIRWISE TRAINING OVERLAP REPORT
# ============================================================

print("\nCreating pairwise overlap report...")

overlap_rows = []

for i in range(len(model_specs)):

    name_i = model_specs[i]["model_name"]
    set_i = sample_sets[name_i]

    for j in range(i + 1, len(model_specs)):

        name_j = model_specs[j]["model_name"]
        set_j = sample_sets[name_j]

        intersection = len(
            set_i.intersection(set_j)
        )

        union = len(
            set_i.union(set_j)
        )

        overlap_rows.append({

            "model_1": name_i,

            "model_2": name_j,

            "group_1": model_specs[i]["group"],

            "group_2": model_specs[j]["group"],

            "overlap_rows": intersection,

            "overlap_fraction_smaller": (
                intersection / min(len(set_i), len(set_j))
            ),

            "jaccard_overlap": (
                intersection / union
                if union > 0 else 0.0
            )

        })


overlap_df = pd.DataFrame(overlap_rows)

overlap_df.to_csv(
    OUTPUT_DIR / "pairwise_train_overlap.csv",
    index=False
)

print("Pairwise overlap rows:", len(overlap_df))


# ============================================================
# 8. ROOT SPLIT EXTRACTION
# ============================================================

def extract_root_vector(json_path):

    with open(
        json_path,
        "r",
        encoding="utf-8"
    ) as f:

        model_json = json.load(f)

    trees = model_json.get("oblivious_trees")

    if trees is None:
        raise RuntimeError(
            f"Missing oblivious_trees in {json_path}"
        )

    if len(trees) != CB_ITERATIONS:
        raise RuntimeError(
            f"Expected {CB_ITERATIONS} trees, "
            f"found {len(trees)} in {json_path}"
        )

    # 0 = Numerical
    # 1 = Categorical
    # 2 = No split

    vector = np.full(
        CB_ITERATIONS,
        2,
        dtype=np.uint8
    )

    audit = []

    for tree_idx, tree in enumerate(trees):

        splits = tree.get("splits") or []

        # Tree with no split.
        if len(splits) == 0:

            vector[tree_idx] = 2

            audit.append({

                "tree_index": tree_idx + 1,

                "root_value": 2,

                "root_kind": "no_split",

                "root_split_type": None,

                "root_split_details": None,

                "has_root_split": False

            })

            continue

        # CatBoost SymmetricTree JSON:
        # Last split entry is treated as the root.
        root_split = splits[-1]

        split_type = root_split.get("split_type")

        if split_type in NUMERICAL_TYPES:

            value = 0
            kind = "numerical"

        elif split_type in CATEGORICAL_TYPES:

            value = 1
            kind = "categorical"

        else:

            raise RuntimeError(
                f"Unknown root split type: {split_type}; "
                f"tree {tree_idx + 1}; file {json_path}"
            )

        vector[tree_idx] = value

        audit.append({

            "tree_index": tree_idx + 1,

            "root_value": value,

            "root_kind": kind,

            "root_split_type": split_type,

            "root_split_details": json.dumps(
                root_split,
                ensure_ascii=False
            ),

            "has_root_split": True

        })

    assert vector.shape == (CB_ITERATIONS,)

    assert np.isin(vector, [0, 1, 2]).all()

    assert len(audit) == CB_ITERATIONS

    return vector, audit


# ============================================================
# 9. TRAIN THE 20 CATBOOST MODELS
# ============================================================

print("\n" + "=" * 75)
print("TRAINING 20 CATBOOST MODELS")
print("=" * 75)

all_vectors = []
all_audits = []
vector_summary = []

for model_number, spec in enumerate(
    model_specs,
    start=1
):

    model_name = spec["model_name"]
    fit_idx = spec["fit_idx"]

    print("\n" + "-" * 75)

    print(
        f"MODEL {model_number}/{EXPECTED_MODELS}: "
        f"{model_name}"
    )

    print("Group:", spec["group"])

    print("Training rows:", len(fit_idx))

    print(
        "Non-default:",
        int(np.sum(y_array[fit_idx] == 0))
    )

    print(
        "Default:",
        int(np.sum(y_array[fit_idx] == 1))
    )

    train_pool = Pool(
        X_cb.iloc[fit_idx],
        y_array[fit_idx],
        cat_features=CAT_FEATURES
    )

    cb_model = CatBoostClassifier(

        iterations=CB_ITERATIONS,

        depth=CB_DEPTH,

        learning_rate=CB_LEARNING_RATE,

        l2_leaf_reg=CB_L2_LEAF_REG,

        loss_function="Logloss",

        eval_metric="AUC",

        random_seed=int(spec["seed"]),

        verbose=CB_VERBOSE_VALUE,

        allow_writing_files=False,

        thread_count=-1,

        grow_policy="SymmetricTree"

    )

    # Train exactly CB_ITERATIONS trees.
    cb_model.fit(train_pool)

    if cb_model.tree_count_ != CB_ITERATIONS:

        raise RuntimeError(
            f"{model_name}: expected {CB_ITERATIONS} trees, "
            f"got {cb_model.tree_count_}"
        )

    # Save native CatBoost model.
    model_path = MODEL_DIR / f"{model_name}.cbm"

    cb_model.save_model(
        str(model_path),
        format="cbm"
    )

    # Export JSON.
    json_path = MODEL_DIR / f"{model_name}.json"

    cb_model.save_model(
        str(json_path),
        format="json"
    )

    # Extract root vector.
    vector, audit = extract_root_vector(json_path)

    for row in audit:

        row["model_name"] = model_name
        row["group"] = spec["group"]

    all_vectors.append(vector)

    all_audits.extend(audit)

    # Save individual NPY vector.
    np.save(
        INDIVIDUAL_DIR / f"{model_name}_vector.npy",
        vector
    )

    # Save individual CSV vector.
    vector_columns = [
        f"TREE_{i:03d}"
        for i in range(1, CB_ITERATIONS + 1)
    ]

    pd.DataFrame(
        [vector],
        columns=vector_columns
    ).to_csv(
        INDIVIDUAL_DIR / f"{model_name}_vector.csv",
        index=False
    )

    # Count root types.
    n_num = int(np.sum(vector == 0))
    n_cat = int(np.sum(vector == 1))
    n_no_split = int(np.sum(vector == 2))

    vector_summary.append({

        "model_name": model_name,

        "group": spec["group"],

        "train_rows": len(fit_idx),

        "numerical_root_count": n_num,

        "categorical_root_count": n_cat,

        "no_split_tree_count": n_no_split,

        "numerical_fraction": n_num / CB_ITERATIONS,

        "categorical_fraction": n_cat / CB_ITERATIONS,

        "no_split_fraction": n_no_split / CB_ITERATIONS

    })

    print("Model saved:", model_path)

    print("Vector shape:", vector.shape)

    print("Numerical roots:", n_num)

    print("Categorical roots:", n_cat)

    print("Trees without splits:", n_no_split)

    print("Total:", n_num + n_cat + n_no_split)

    del train_pool
    del cb_model

    gc.collect()


# ============================================================
# 10. SAVE COMBINED ROOT VECTORS
# ============================================================

print("\nSaving the 20 root vectors...")

root_vectors = np.stack(
    all_vectors,
    axis=0
).astype(np.uint8)

if root_vectors.shape != (20, CB_ITERATIONS):

    raise RuntimeError(
        f"Expected (20, {CB_ITERATIONS}), "
        f"got {root_vectors.shape}"
    )

if not np.isin(root_vectors, [0, 1, 2]).all():

    raise RuntimeError(
        "Invalid values found in root_vectors."
    )

np.save(
    VECTOR_DIR / "root_vectors.npy",
    root_vectors
)

vector_columns = [
    f"TREE_{i:03d}"
    for i in range(1, CB_ITERATIONS + 1)
]

root_df = pd.DataFrame(
    root_vectors,
    columns=vector_columns
)

root_df.insert(
    0,
    "model_name",
    [s["model_name"] for s in model_specs]
)

root_df.insert(
    1,
    "group",
    [s["group"] for s in model_specs]
)

root_df.to_csv(
    VECTOR_DIR / "root_vectors.csv",
    index=False
)


# ============================================================
# 11. SAVE TREE-LEVEL AUDIT
# ============================================================

audit_df = pd.DataFrame(all_audits)

audit_df.to_csv(
    OUTPUT_DIR / "root_split_audit.csv",
    index=False
)


# ============================================================
# 12. SAVE VECTOR SUMMARY AND MODEL ORDER
# ============================================================

summary_vectors_df = pd.DataFrame(
    vector_summary
)

summary_vectors_df.to_csv(
    VECTOR_DIR / "root_vector_summary.csv",
    index=False
)

print("\nRoot vector summary:")

print(
    summary_vectors_df[
        [
            "model_name",
            "group",
            "numerical_root_count",
            "categorical_root_count",
            "no_split_tree_count"
        ]
    ].to_string(index=False)
)

order_df = pd.DataFrame({

    "matrix_row": np.arange(1, 21),

    "model_name": [
        s["model_name"] for s in model_specs
    ],

    "group": [
        s["group"] for s in model_specs
    ]

})

order_df.to_csv(
    VECTOR_DIR / "root_vector_model_order.csv",
    index=False
)


# ============================================================
# 13. SAVE EXPERIMENT METADATA
# ============================================================

group_B_metadata = []

for spec in group_B_specs:

    group_B_metadata.append({

        "model_name": spec["model_name"],

        "paired_A_model": spec["paired_A_model"],

        "default_count": spec["positive_count"],

        "non_default_count": spec["negative_count"],

        "total_training_rows": len(spec["fit_idx"]),

        "ratio": "3:1"

    })


metadata = {

    "experiment": "FULLPOS_3TO1",

    "total_models": 20,

    "group_a_models": 10,

    "group_b_models": 10,

    "trees_per_model": CB_ITERATIONS,

    "root_vector_shape": [
        20,
        CB_ITERATIONS
    ],

    "encoding": {

        "0": "Numerical root split",

        "1": "Categorical root split",

        "2": "Tree without split"

    },

    "categorical_split_types": sorted(
        CATEGORICAL_TYPES
    ),

    "numerical_split_types": sorted(
        NUMERICAL_TYPES
    ),

    "catboost_parameters": {

        "iterations": CB_ITERATIONS,

        "depth": CB_DEPTH,

        "learning_rate": CB_LEARNING_RATE,

        "l2_leaf_reg": CB_L2_LEAF_REG,

        "loss_function": "Logloss",

        "eval_metric": "AUC",

        "grow_policy": "SymmetricTree",

        "class_weights": None

    },

    "group_a_sampling": (
        "Original two repetitions of StratifiedKFold, "
        "five folds per repetition."
    ),

    "group_b_sampling": {

        "method": (
            "For every corresponding A model, retain all "
            "default samples from that A training split and "
            "randomly select three times as many non-default "
            "samples from the same split."
        ),

        "negative_to_positive_ratio": "3:1",

        "positive_rows_repeated_between_models": True,

        "negative_rows_repeated_between_models": True,

        "sampling_without_replacement_within_each_model": True,

        "paired_models": group_B_metadata

    },

    "cross_group_overlap_allowed": True,

    "seed": SEED_VALUE,

    "model_order": [
        s["model_name"] for s in model_specs
    ]

}

with open(
    OUTPUT_DIR / "experiment_metadata.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=2,
        ensure_ascii=False
    )


# ============================================================
# 14. FINAL VALIDATION
# ============================================================

print("\nFinal validation...")

cbm_files = list(MODEL_DIR.glob("*.cbm"))
json_files = list(MODEL_DIR.glob("*.json"))
npy_files = list(INDIVIDUAL_DIR.glob("*_vector.npy"))

assert len(cbm_files) == 20
assert len(json_files) == 20
assert len(npy_files) == 20

assert root_vectors.shape == (20, CB_ITERATIONS)

assert np.isin(root_vectors, [0, 1, 2]).all()

assert len(all_audits) == 20 * CB_ITERATIONS

assert len(overlap_df) == 190

assert len(group_A_specs) == 10
assert len(group_B_specs) == 10

for i in range(20):

    assert len(root_vectors[i]) == CB_ITERATIONS

    assert (
        np.sum(root_vectors[i] == 0)
        + np.sum(root_vectors[i] == 1)
        + np.sum(root_vectors[i] == 2)
    ) == CB_ITERATIONS


# Validate B sampling.
for spec in group_B_specs:

    idx = spec["fit_idx"]

    n_positive = int(
        np.sum(y_array[idx] == 1)
    )

    n_negative = int(
        np.sum(y_array[idx] == 0)
    )

    assert n_negative == 3 * n_positive


# ============================================================
# 15. FINAL REPORT
# ============================================================

print("\n" + "=" * 75)
print("EXPERIMENT 2 COMPLETED SUCCESSFULLY")
print("=" * 75)

print("CatBoost models:", len(cbm_files))
print("JSON models:", len(json_files))
print("Individual vectors:", len(npy_files))

print("Combined vector shape:", root_vectors.shape)

print("\nTraining groups:")

print(
    sampling_summary_df[
        sampling_summary_df["group"] == "A_original"
    ][
        ["training_rows", "non_default", "default"]
    ].describe().round(2)
)

print(
    "\nGroup B sampling:"
)

print(
    sampling_summary_df[
        sampling_summary_df["group"] == "B_ratio3to1"
    ][
        ["training_rows", "non_default", "default"]
    ].to_string(index=False)
)

print("\nEncoding:")
print("0 = Numerical")
print("1 = Categorical")
print("2 = No split")

print(
    "\nTotal numerical roots:",
    int(np.sum(root_vectors == 0))
)

print(
    "Total categorical roots:",
    int(np.sum(root_vectors == 1))
)

print(
    "Total trees without splits:",
    int(np.sum(root_vectors == 2))
)

print("\nMain output directory:")
print(OUTPUT_DIR)

print("\nVector matrix:")
print(VECTOR_DIR / "root_vectors.npy")

print("\nReadable vectors:")
print(VECTOR_DIR / "root_vectors.csv")

print("\nRoot split audit:")
print(OUTPUT_DIR / "root_split_audit.csv")

print("\nTraining sample manifest:")
print(OUTPUT_DIR / "training_sample_manifest.csv")

print("\nPairwise overlap:")
print(OUTPUT_DIR / "pairwise_train_overlap.csv")

print("\nExperiment metadata:")
print(OUTPUT_DIR / "experiment_metadata.json")

print("\n" + "=" * 75)
print("ALL 20 MODELS AND ROOT VECTORS SAVED SUCCESSFULLY")
print("=" * 75)
```

```python

# ============================================================
# 15. TEST PREDICTIONS + 20 INDIVIDUAL SUBMISSIONS + ENSEMBLES
# EXPERIMENT: FULL DEFAULT RETENTION + 3:1
# HOME CREDIT - 20 CATBOOST MODELS
# ============================================================

import re
import gc
import numpy as np
import pandas as pd

from pathlib import Path
from catboost import CatBoostClassifier


print("\n" + "=" * 75)
print("GENERATING TEST SUBMISSIONS - FULLPOS_3TO1 EXPERIMENT")
print("=" * 75)


# ============================================================
# 1. CONFIGURATION - NEW EXPERIMENT PATH
# ============================================================

MODEL_ROOT = Path(
    "/content/drive/MyDrive/Competition/output/"
    "CATBOOST_20_ROOT_VECTORS_FULLPOS_3TO1"
)

MODELS_DIR = MODEL_ROOT / "models"

SUBMISSION_DIR = MODEL_ROOT / "test_submissions"

SUBMISSION_DIR.mkdir(
    parents=True,
    exist_ok=True
)

ID_COL = "SK_ID_CURR"
TARGET_COL = "TARGET"

EXPECTED_MODELS = 20


# ============================================================
# 2. GET ORIGINAL TEST DATA
# ============================================================

if "test" in globals():

    test_df = test.copy()

elif "test_df" in globals():

    test_df = test_df.copy()

else:

    raise NameError(
        "Neither 'test' nor 'test_df' exists. "
        "Load competition test.csv first."
    )


if ID_COL not in test_df.columns:

    raise ValueError(
        f"Test dataset is missing {ID_COL}."
    )


if TARGET_COL in test_df.columns:

    raise ValueError(
        "TARGET must not exist in the competition test dataset."
    )


# ============================================================
# 3. CREATE CREDIT_TERM
# ============================================================

# Same feature definition used in the previous
# submission-generation experiment:
#
# CREDIT_TERM = AMT_CREDIT / AMT_ANNUITY

required_columns = [
    "AMT_CREDIT",
    "AMT_ANNUITY"
]

missing_columns = [
    col for col in required_columns
    if col not in test_df.columns
]

if missing_columns:

    raise ValueError(
        "Cannot create CREDIT_TERM. "
        f"Missing columns: {missing_columns}"
    )


credit = pd.to_numeric(
    test_df["AMT_CREDIT"],
    errors="coerce"
)

annuity = pd.to_numeric(
    test_df["AMT_ANNUITY"],
    errors="coerce"
)

annuity = annuity.where(
    annuity != 0,
    np.nan
)

test_df["CREDIT_TERM"] = (
    credit / annuity
)

test_df["CREDIT_TERM"] = (
    test_df["CREDIT_TERM"]
    .replace(
        [np.inf, -np.inf],
        np.nan
    )
)


print("\nCREDIT_TERM successfully created.")

print(
    "CREDIT_TERM missing values:",
    test_df["CREDIT_TERM"].isna().sum()
)

print(
    "CREDIT_TERM mean:",
    test_df["CREDIT_TERM"].mean()
)

print(f"\nTest rows: {len(test_df):,}")
print(f"Test columns: {len(test_df.columns)}")


# ============================================================
# 4. VALIDATE NEW MODEL DIRECTORY
# ============================================================

if not MODELS_DIR.exists():

    raise FileNotFoundError(
        f"New experiment models directory not found:\n"
        f"{MODELS_DIR}"
    )


model_paths = list(
    MODELS_DIR.glob("*.cbm")
)


def natural_sort_key(path):

    return [
        int(part) if part.isdigit() else part.lower()
        for part in re.split(r"(\d+)", path.stem)
    ]


model_paths = sorted(
    model_paths,
    key=natural_sort_key
)


if len(model_paths) != EXPECTED_MODELS:

    raise RuntimeError(
        f"Expected {EXPECTED_MODELS} new CatBoost models, "
        f"but found {len(model_paths)}.\n"
        f"Directory: {MODELS_DIR}"
    )


print(f"\nFound {len(model_paths)} CatBoost models.")


# ============================================================
# 5. VALIDATE MODEL NAMES AND GROUPS
# ============================================================

model_names = [
    path.stem
    for path in model_paths
]

group_A_names = [
    name for name in model_names
    if name.startswith("A_original_")
]

group_B_names = [
    name for name in model_names
    if name.startswith("B_ratio3to1_")
]


if len(group_A_names) != 10:

    raise RuntimeError(
        f"Expected 10 A_original models, "
        f"found {len(group_A_names)}."
    )


if len(group_B_names) != 10:

    raise RuntimeError(
        f"Expected 10 B_ratio3to1 models, "
        f"found {len(group_B_names)}."
    )


unknown_models = [
    name for name in model_names
    if (
        not name.startswith("A_original_")
        and not name.startswith("B_ratio3to1_")
    )
]


if unknown_models:

    raise RuntimeError(
        f"Unknown model names detected: {unknown_models}"
    )


print("\nModel groups validated:")
print("Group A:", len(group_A_names))
print("Group B:", len(group_B_names))


# ============================================================
# 6. PRE-VALIDATE FEATURES FOR ALL 20 MODELS
# ============================================================

print("\nPre-validating all model features...")


model_feature_specs = {}


for model_path in model_paths:

    model_name = model_path.stem

    model = CatBoostClassifier()

    model.load_model(
        str(model_path)
    )

    feature_names = model.feature_names_

    if not feature_names:

        raise RuntimeError(
            f"No stored feature names in {model_name}."
        )


    missing_features = [
        feature
        for feature in feature_names
        if feature not in test_df.columns
    ]


    if missing_features:

        raise ValueError(
            f"\nMissing features for {model_name}:\n"
            f"{missing_features}"
        )


    model_feature_specs[model_name] = feature_names

    print(
        f"Validated: {model_name} "
        f"({len(feature_names)} features)"
    )

    del model


gc.collect()


print(
    "\nAll 20 models passed feature validation."
)


# ============================================================
# 7. PREDICT WITH EACH MODEL
# ============================================================

all_predictions = {}

model_groups = {}


for model_idx, model_path in enumerate(
    model_paths,
    start=1
):

    model_name = model_path.stem

    print("\n" + "-" * 70)

    print(
        f"MODEL {model_idx}/{EXPECTED_MODELS}: "
        f"{model_name}"
    )


    # --------------------------------------------------------
    # Load saved model
    # --------------------------------------------------------

    model = CatBoostClassifier()

    model.load_model(
        str(model_path)
    )


    # --------------------------------------------------------
    # Select exact training feature order
    # --------------------------------------------------------

    feature_names = model_feature_specs[model_name]

    X_test = test_df[feature_names]


    # --------------------------------------------------------
    # Identify positive class TARGET = 1
    # --------------------------------------------------------

    classes = list(model.classes_)

    positive_class_indices = [

        i for i, cls in enumerate(classes)

        if cls == 1 or str(cls) == "1"

    ]


    if len(positive_class_indices) != 1:

        raise RuntimeError(
            f"Could not identify TARGET=1 for "
            f"{model_name}. Classes: {classes}"
        )


    positive_class_idx = positive_class_indices[0]


    # --------------------------------------------------------
    # Predict probabilities
    # --------------------------------------------------------

    probabilities = model.predict_proba(
        X_test
    )[:, positive_class_idx]


    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )


    # --------------------------------------------------------
    # Validate predictions
    # --------------------------------------------------------

    if len(probabilities) != len(test_df):

        raise RuntimeError(
            f"Prediction length mismatch: {model_name}"
        )


    if not np.isfinite(probabilities).all():

        raise RuntimeError(
            f"Invalid predictions detected: {model_name}"
        )


    if (
        probabilities.min() < 0
        or probabilities.max() > 1
    ):

        raise RuntimeError(
            f"Probabilities outside [0, 1]: {model_name}"
        )


    # --------------------------------------------------------
    # Store predictions
    # --------------------------------------------------------

    all_predictions[model_name] = probabilities


    # --------------------------------------------------------
    # Identify group
    # --------------------------------------------------------

    if model_name.startswith("A_original_"):

        model_groups[model_name] = "A_original"

    elif model_name.startswith("B_ratio3to1_"):

        model_groups[model_name] = "B_ratio3to1"


    # --------------------------------------------------------
    # Save individual submission
    # --------------------------------------------------------

    individual_submission = pd.DataFrame({

        ID_COL: test_df[ID_COL].values,

        TARGET_COL: probabilities

    })


    individual_path = (
        SUBMISSION_DIR /
        f"submission_{model_name}.csv"
    )


    individual_submission.to_csv(
        individual_path,
        index=False
    )


    print(
        f"Saved: {individual_path.name}"
    )

    print(
        f"Mean: {probabilities.mean():.6f} | "
        f"Std: {probabilities.std():.6f} | "
        f"Min: {probabilities.min():.6f} | "
        f"Max: {probabilities.max():.6f}"
    )


    del model
    del X_test
    del individual_submission

    gc.collect()


# ============================================================
# 8. BUILD COMPLETE PREDICTION MATRIX
# ============================================================

prediction_matrix = pd.DataFrame({

    ID_COL: test_df[ID_COL].values

})


for model_name in model_names:

    prediction_matrix[model_name] = (
        all_predictions[model_name]
    )


# ============================================================
# 9. ENSEMBLE A - ORIGINAL DISTRIBUTION
# ============================================================

prediction_matrix["ENSEMBLE_A_ORIGINAL"] = (
    prediction_matrix[group_A_names].mean(axis=1)
)


submission_A = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_A_ORIGINAL"
    ].values

})


submission_A.to_csv(
    SUBMISSION_DIR / "submission_ensemble_A_original.csv",
    index=False
)


# ============================================================
# 10. ENSEMBLE B - FULL DEFAULT RETENTION + 3:1
# ============================================================

prediction_matrix["ENSEMBLE_B_FULLPOS_3TO1"] = (
    prediction_matrix[group_B_names].mean(axis=1)
)


submission_B = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_B_FULLPOS_3TO1"
    ].values

})


submission_B.to_csv(
    SUBMISSION_DIR / "submission_ensemble_B_FULLPOS_3TO1.csv",
    index=False
)


# ============================================================
# 11. ENSEMBLE ALL 20 MODELS
# ============================================================

all_model_columns = (
    group_A_names + group_B_names
)


prediction_matrix["ENSEMBLE_MEAN_20"] = (
    prediction_matrix[all_model_columns].mean(axis=1)
)


submission_20 = pd.DataFrame({

    ID_COL: test_df[ID_COL].values,

    TARGET_COL: prediction_matrix[
        "ENSEMBLE_MEAN_20"
    ].values

})


submission_20.to_csv(
    SUBMISSION_DIR / "submission_ensemble_20.csv",
    index=False
)


# ============================================================
# 12. MODEL DISAGREEMENT
# ============================================================

prediction_matrix["MODEL_PREDICTION_STD"] = (
    prediction_matrix[all_model_columns].std(
        axis=1,
        ddof=0
    )
)


prediction_matrix.to_csv(
    SUBMISSION_DIR / "all_20_model_predictions.csv",
    index=False
)


# ============================================================
# 13. SAVE TEST PREDICTION SUMMARY
# ============================================================

summary_rows = []


for model_name, probabilities in all_predictions.items():

    summary_rows.append({

        "model_name": model_name,

        "group": model_groups[model_name],

        "prediction_mean": probabilities.mean(),

        "prediction_std": probabilities.std(),

        "prediction_min": probabilities.min(),

        "prediction_max": probabilities.max()

    })


summary_df = pd.DataFrame(
    summary_rows
)


summary_df.to_csv(
    SUBMISSION_DIR / "test_prediction_summary.csv",
    index=False
)


# ============================================================
# 14. FINAL VALIDATION
# ============================================================

submission_files = [

    SUBMISSION_DIR / f"submission_{name}.csv"

    for name in model_names

]


submission_files.extend([

    SUBMISSION_DIR / "submission_ensemble_A_original.csv",

    SUBMISSION_DIR / "submission_ensemble_B_FULLPOS_3TO1.csv",

    SUBMISSION_DIR / "submission_ensemble_20.csv"

])


for file_path in submission_files:

    if not file_path.exists():

        raise RuntimeError(
            f"Expected submission not found: {file_path.name}"
        )

    check = pd.read_csv(file_path)

    if list(check.columns) != [ID_COL, TARGET_COL]:

        raise RuntimeError(
            f"Invalid submission columns: {file_path.name}"
        )

    if len(check) != len(test_df):

        raise RuntimeError(
            f"Invalid submission row count: {file_path.name}"
        )

    if not np.isfinite(
        check[TARGET_COL].to_numpy()
    ).all():

        raise RuntimeError(
            f"Invalid TARGET probabilities: {file_path.name}"
        )

    if check[ID_COL].tolist() != test_df[ID_COL].tolist():

        raise RuntimeError(
            f"ID order mismatch: {file_path.name}"
        )


# ============================================================
# 15. FINAL REPORT
# ============================================================

print("\n" + "=" * 75)
print("FULLPOS_3TO1 SUBMISSION GENERATION COMPLETED")
print("=" * 75)


print(f"\nTest rows: {len(test_df):,}")

print(
    "Individual model submissions:",
    len(all_predictions)
)


print("\nGROUP A: ORIGINAL DISTRIBUTION")

print("Models:", len(group_A_names))

print(
    "Mean prediction:",
    f"{submission_A[TARGET_COL].mean():.6f}"
)


print("\nGROUP B: FULL DEFAULT RETENTION + 3:1")

print("Models:", len(group_B_names))

print(
    "Mean prediction:",
    f"{submission_B[TARGET_COL].mean():.6f}"
)


print("\nENSEMBLE OF ALL 20 MODELS")

print(
    "Mean prediction:",
    f"{submission_20[TARGET_COL].mean():.6f}"
)


print(
    "Prediction disagreement mean:",
    f"{prediction_matrix['MODEL_PREDICTION_STD'].mean():.6f}"
)


print("\nOUTPUT DIRECTORY:")

print(SUBMISSION_DIR)


print("\nGENERATED FILES:")

print("- 10 A_original individual submissions")

print("- 10 B_ratio3to1 individual submissions")

print("- submission_ensemble_A_original.csv")

print("- submission_ensemble_B_FULLPOS_3TO1.csv")

print("- submission_ensemble_20.csv")

print("- all_20_model_predictions.csv")

print("- test_prediction_summary.csv")


print("\nAll submissions saved and validated successfully.")

print("=" * 75)
```

```python

# ============================================================
# HOME CREDIT - B + ALL COMBINATIONS OF 6 NEW FEATURES
#
# 64 COMBINATIONS:
# C00 = CREDIT_TERM baseline
# C01-C63 = all non-empty subsets of six optional features
#
# 10 B MODELS PER COMBINATION = 640 MODELS
#
# EARLY STOPPING:
# Maximum iterations = 5000
# Early stopping rounds = 200
#
# SAVED:
# - 640 CatBoost CBM models
# - 640 CatBoost JSON models
# - 640 variable-length root vectors
# - Root split audits
# - 640 validation and test prediction checkpoints
# - Complete test prediction matrix for future blending
# - 64 combination submissions
# - Best submission based on pooled repeated OOF AUC
# - CV metrics and experiment metadata
#
# Original FULLPOS_3TO1 experiment remains untouched.
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

import os
import gc
import re
import json
import gzip
import hashlib
import warnings

from pathlib import Path
from itertools import combinations
from time import perf_counter

import numpy as np
import pandas as pd

from catboost import CatBoostClassifier

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    log_loss,
    brier_score_loss
)

warnings.filterwarnings("ignore")


# ============================================================
# 1. PATHS AND CONFIGURATION
# ============================================================

BASE_EXPERIMENT_DIR = Path(
    "/content/drive/MyDrive/Competition/output/"
    "CATBOOST_20_ROOT_VECTORS_FULLPOS_3TO1"
)

TRAIN_FILE = Path(
    "/content/drive/MyDrive/Competition/input/train.csv"
)

TEST_FILE = Path(
    "/content/drive/MyDrive/Competition/input/test.csv"
)

MANIFEST_FILE = (
    BASE_EXPERIMENT_DIR /
    "training_sample_manifest.csv"
)

BASE_METADATA_FILE = (
    BASE_EXPERIMENT_DIR /
    "experiment_metadata.json"
)

BASE_MODEL_FILE = (
    BASE_EXPERIMENT_DIR /
    "models" /
    "B_ratio3to1_01.cbm"
)


# Completely separate experiment directory.
OUTPUT_DIR = Path(
    "/content/drive/MyDrive/Competition/output/"
    "CATBOOST_B_6FEATURES_ALL_COMBINATIONS"
)

MODEL_DIR = OUTPUT_DIR / "models"

PREDICTION_CHECKPOINT_DIR = (
    OUTPUT_DIR / "prediction_checkpoints"
)

VECTOR_DIR = OUTPUT_DIR / "root_vectors"

VECTOR_INDIVIDUAL_DIR = (
    VECTOR_DIR / "individual"
)

AUDIT_DIR = OUTPUT_DIR / "tree_audits"

SUBMISSION_DIR = OUTPUT_DIR / "submissions"

for directory in [
    OUTPUT_DIR,
    MODEL_DIR,
    PREDICTION_CHECKPOINT_DIR,
    VECTOR_DIR,
    VECTOR_INDIVIDUAL_DIR,
    AUDIT_DIR,
    SUBMISSION_DIR
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ------------------------------------------------------------
# Training configuration
# ------------------------------------------------------------

MAX_ITERATIONS = 5000

EARLY_STOPPING_ROUNDS = 200

LOG_PERIOD = 200

TASK_TYPE = "CPU"

MISSING_TOKEN = "__MISSING__"

B_MODEL_COUNT = 10

EXPECTED_COMBINATIONS = 64

EXPECTED_TOTAL_MODELS = 640

ID_COL = "SK_ID_CURR"

TARGET_COL = "TARGET"

SEED_OFFSET_B = 20000


# Six optional features.
# CREDIT_TERM is already present in the 33-feature base matrix.

NEW_FEATURES = [

    "ID_PUBLISH_AGE",

    "AGE_YEARS",

    "INCOME_TO_REGION_MEAN",

    "INCOME_GAP_FROM_REGION_MEAN",

    "WORKING_CASH_INDICATOR",

    "WORKING_CASH_EMPLOYMENT_YEARS"

]


# ============================================================
# 2. VALIDATE PREVIOUS EXPERIMENT ARTIFACTS
# ============================================================

print("\n" + "=" * 85)
print("VALIDATING PREVIOUS FULLPOS_3TO1 EXPERIMENT")
print("=" * 85)


required_paths = [
    TRAIN_FILE,
    TEST_FILE,
    MANIFEST_FILE,
    BASE_METADATA_FILE,
    BASE_MODEL_FILE
]

missing_paths = [
    str(path)
    for path in required_paths
    if not path.exists()
]

if missing_paths:

    raise FileNotFoundError(
        "Required previous-experiment files are missing:\n"
        + "\n".join(missing_paths)
    )


with open(
    BASE_METADATA_FILE,
    "r",
    encoding="utf-8"
) as f:

    base_metadata = json.load(f)


with open(
    BASE_METADATA_FILE,
    "r",
    encoding="utf-8"
) as f:

    base_metadata_check = json.load(f)


# Recover original CatBoost settings.
saved_cb_params = base_metadata["catboost_parameters"]

DEPTH = int(saved_cb_params["depth"])

LEARNING_RATE = float(
    saved_cb_params["learning_rate"]
)

L2_LEAF_REG = float(
    saved_cb_params["l2_leaf_reg"]
)

BASE_SEED = int(
    base_metadata.get("seed", 42)
)

GROW_POLICY = saved_cb_params.get(
    "grow_policy",
    "SymmetricTree"
)


print("Previous experiment:", BASE_EXPERIMENT_DIR)
print("Original depth:", DEPTH)
print("Original learning rate:", LEARNING_RATE)
print("Original L2:", L2_LEAF_REG)
print("Original seed:", BASE_SEED)

print("Maximum iterations:", MAX_ITERATIONS)
print("Early stopping:", EARLY_STOPPING_ROUNDS)


# ============================================================
# 3. LOAD RAW DATA
# ============================================================

print("\nLoading train and test data...")

train_raw_df = pd.read_csv(
    TRAIN_FILE,
    low_memory=False
)

test_raw_df = pd.read_csv(
    TEST_FILE,
    low_memory=False
)

print("Train shape:", train_raw_df.shape)
print("Test shape:", test_raw_df.shape)


if ID_COL not in train_raw_df.columns:
    raise ValueError("SK_ID_CURR is missing from train.")

if TARGET_COL not in train_raw_df.columns:
    raise ValueError("TARGET is missing from train.")

if ID_COL not in test_raw_df.columns:
    raise ValueError("SK_ID_CURR is missing from test.")

if TARGET_COL in test_raw_df.columns:
    raise ValueError("TARGET must not be present in test.")


# Reset positions to ensure manifest row indices remain positional.
train_raw_df = train_raw_df.reset_index(drop=True)

test_raw_df = test_raw_df.reset_index(drop=True)

y_array = pd.to_numeric(
    train_raw_df[TARGET_COL],
    errors="raise"
).astype(np.uint8).to_numpy()

test_ids = test_raw_df[ID_COL].to_numpy()

N_TRAIN = len(train_raw_df)

N_TEST = len(test_raw_df)

if not np.isin(y_array, [0, 1]).all():
    raise ValueError("TARGET must contain only 0 and 1.")


print("Training rows:", N_TRAIN)
print("Test rows:", N_TEST)

print("Non-default:", int(np.sum(y_array == 0)))
print("Default:", int(np.sum(y_array == 1)))


# ============================================================
# 4. LOAD ORIGINAL BASE CATBOOST MODEL
# ============================================================

print("\nLoading original B model for feature schema...")

base_model = CatBoostClassifier()

base_model.load_model(
    str(BASE_MODEL_FILE)
)

BASE_FEATURES = list(
    base_model.feature_names_
)

if not BASE_FEATURES:
    raise RuntimeError(
        "Original B model does not contain feature names."
    )

if "CREDIT_TERM" not in BASE_FEATURES:
    raise RuntimeError(
        "CREDIT_TERM is not present in the original B model."
    )

if len(BASE_FEATURES) != 33:
    raise RuntimeError(
        f"Expected 33 base features, found {len(BASE_FEATURES)}."
    )


cat_indices = base_model.get_cat_feature_indices()

CAT_FEATURES = [
    BASE_FEATURES[i]
    for i in cat_indices
]

print("Base feature count:", len(BASE_FEATURES))
print("Categorical feature count:", len(CAT_FEATURES))
print("Categorical features:", CAT_FEATURES)


# ============================================================
# 5. BUILD BASE TRAINING MATRIX
# ============================================================

# Use the original X_cb matrix whenever available.
# This preserves the exact training preprocessing.

if "X_cb" in globals():

    print("\nUsing existing X_cb training matrix.")

    X_base_train = X_cb.copy().reset_index(drop=True)

else:

    print(
        "\nX_cb is unavailable. Reconstructing the base matrix "
        "from raw training data."
    )

    X_base_train = train_raw_df.drop(
        columns=[TARGET_COL, ID_COL]
    ).copy()

    credit = pd.to_numeric(
        X_base_train["AMT_CREDIT"],
        errors="coerce"
    )

    annuity = pd.to_numeric(
        X_base_train["AMT_ANNUITY"],
        errors="coerce"
    )

    annuity = annuity.where(
        annuity != 0,
        np.nan
    )

    X_base_train["CREDIT_TERM"] = (
        credit / annuity
    )

    X_base_train = X_base_train.reindex(
        columns=BASE_FEATURES
    )


if len(X_base_train) != N_TRAIN:
    raise ValueError(
        "X_cb and raw train have different row counts."
    )

if set(X_base_train.columns) != set(BASE_FEATURES):
    raise ValueError(
        "X_cb feature schema does not match the original B model."
    )

X_base_train = X_base_train.reindex(
    columns=BASE_FEATURES
).reset_index(drop=True)


# Validate target alignment if global y exists.
if "y" in globals():

    y_existing = np.asarray(y).reshape(-1)

    if (
        len(y_existing) != N_TRAIN
        or not np.array_equal(
            y_existing.astype(int),
            y_array.astype(int)
        )
    ):
        raise ValueError(
            "Existing y does not match train.csv row order."
        )


# ============================================================
# 6. BUILD BASE TEST MATRIX
# ============================================================

raw_train_columns = [
    col
    for col in BASE_FEATURES
    if col != "CREDIT_TERM"
]

missing_test_features = [
    col
    for col in raw_train_columns
    if col not in test_raw_df.columns
]

if missing_test_features:
    raise ValueError(
        f"Missing raw test features: {missing_test_features}"
    )


X_base_test = test_raw_df[
    raw_train_columns
].copy()


# CREDIT_TERM = AMT_CREDIT / AMT_ANNUITY

credit = pd.to_numeric(
    X_base_test["AMT_CREDIT"],
    errors="coerce"
)

annuity = pd.to_numeric(
    X_base_test["AMT_ANNUITY"],
    errors="coerce"
)

annuity = annuity.where(
    annuity != 0,
    np.nan
)

X_base_test["CREDIT_TERM"] = (
    credit / annuity
)

X_base_test["CREDIT_TERM"] = (
    X_base_test["CREDIT_TERM"]
    .replace(
        [np.inf, -np.inf],
        np.nan
    )
)

X_base_test = X_base_test.reindex(
    columns=BASE_FEATURES
).reset_index(drop=True)


# Verify training CREDIT_TERM matches the same definition.
expected_credit_term = (
    pd.to_numeric(
        train_raw_df["AMT_CREDIT"],
        errors="coerce"
    )
    /
    pd.to_numeric(
        train_raw_df["AMT_ANNUITY"],
        errors="coerce"
    ).where(
        pd.to_numeric(
            train_raw_df["AMT_ANNUITY"],
            errors="coerce"
        ) != 0,
        np.nan
    )
)

expected_credit_term = expected_credit_term.replace(
    [np.inf, -np.inf],
    np.nan
)

existing_credit_term = pd.to_numeric(
    X_base_train["CREDIT_TERM"],
    errors="coerce"
)

if not np.allclose(
    existing_credit_term.to_numpy(dtype=float),
    expected_credit_term.to_numpy(dtype=float),
    rtol=1e-6,
    atol=1e-8,
    equal_nan=True
):
    raise ValueError(
        "CREDIT_TERM in X_cb does not match "
        "AMT_CREDIT / AMT_ANNUITY."
    )


# ============================================================
# 7. ALIGN CATEGORICAL DATA TYPES
# ============================================================

# New models use the same categorical variables as the original B models.
# The added six features are numerical.

for col in CAT_FEATURES:

    train_col = X_base_train[col]

    test_col = X_base_test[col]

    if (
        pd.api.types.is_object_dtype(train_col.dtype)
        or pd.api.types.is_string_dtype(train_col.dtype)
        or isinstance(
            train_col.dtype,
            pd.CategoricalDtype
        )
    ):

        X_base_train[col] = (
            train_col
            .astype("object")
            .where(train_col.notna(), MISSING_TOKEN)
            .astype(str)
        )

        X_base_test[col] = (
            test_col
            .astype("object")
            .where(test_col.notna(), MISSING_TOKEN)
            .astype(str)
        )

    elif pd.api.types.is_integer_dtype(train_col.dtype):

        if train_col.isna().any():
            raise ValueError(
                f"Numeric categorical feature {col} "
                "contains missing training values."
            )

        numeric_test = pd.to_numeric(
            test_col,
            errors="coerce"
        )

        X_base_test[col] = numeric_test.fillna(-1).astype(np.int64)

    else:

        # Numeric/float categorical inputs are represented as strings.
        X_base_train[col] = (
            train_col
            .astype("object")
            .where(train_col.notna(), MISSING_TOKEN)
            .astype(str)
        )

        X_base_test[col] = (
            test_col
            .astype("object")
            .where(test_col.notna(), MISSING_TOKEN)
            .astype(str)
        )


# Test the feature schema using the original model.
try:

    _ = base_model.predict_proba(
        X_base_test.iloc[:min(10, N_TEST)]
    )

except Exception as exc:

    raise RuntimeError(
        "The constructed test feature matrix is incompatible "
        "with the original CatBoost model. "
        f"Details: {exc}"
    )


print("\nBase train matrix:", X_base_train.shape)
print("Base test matrix:", X_base_test.shape)
print("Categorical alignment validated.")


del base_model
gc.collect()


# ============================================================
# 8. LOAD TRAINING SAMPLE MANIFEST
# ============================================================

print("\nLoading original B training manifest...")

manifest = pd.read_csv(
    MANIFEST_FILE,
    low_memory=False
)

required_manifest_columns = {
    "model_name",
    "group",
    "row_index",
    "TARGET",
    "paired_A_model"
}

missing_manifest_columns = (
    required_manifest_columns - set(manifest.columns)
)

if missing_manifest_columns:
    raise ValueError(
        f"Manifest missing columns: {missing_manifest_columns}"
    )


manifest["row_index"] = pd.to_numeric(
    manifest["row_index"],
    errors="raise"
).astype(np.int64)

if manifest["row_index"].min() < 0:
    raise ValueError("Negative row indices found in manifest.")

if manifest["row_index"].max() >= N_TRAIN:
    raise ValueError("Manifest contains indices outside train.csv.")


# Validate target alignment.
manifest_targets = pd.to_numeric(
    manifest["TARGET"],
    errors="raise"
).astype(int).to_numpy()

manifest_indices = manifest["row_index"].to_numpy()

if not np.array_equal(
    manifest_targets,
    y_array[manifest_indices].astype(int)
):
    raise ValueError(
        "Manifest TARGET values do not match train.csv."
    )


# ============================================================
# 9. RESTORE THE TEN B TRAINING SPLITS AND VALIDATION FOLDS
# ============================================================

print("\nRestoring the ten original B sample sets...")

b_specs = []

all_indices = np.arange(
    N_TRAIN,
    dtype=np.int64
)

for b_idx in range(1, B_MODEL_COUNT + 1):

    b_name = f"B_ratio3to1_{b_idx:02d}"

    b_rows = manifest[
        manifest["model_name"] == b_name
    ]

    if b_rows.empty:
        raise RuntimeError(
            f"Missing B model in manifest: {b_name}"
        )

    b_fit_idx = b_rows["row_index"].to_numpy(
        dtype=np.int64
    )

    paired_names = (
        b_rows["paired_A_model"]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    if len(paired_names) != 1:
        raise RuntimeError(
            f"Invalid paired A model for {b_name}: {paired_names}"
        )

    a_name = paired_names[0]

    a_rows = manifest[
        manifest["model_name"] == a_name
    ]

    if a_rows.empty:
        raise RuntimeError(
            f"Missing paired A model: {a_name}"
        )

    a_fit_idx = a_rows["row_index"].to_numpy(
        dtype=np.int64
    )

    if len(np.unique(b_fit_idx)) != len(b_fit_idx):
        raise RuntimeError(
            f"Duplicate B training rows in {b_name}."
        )

    if not set(b_fit_idx).issubset(set(a_fit_idx)):
        raise RuntimeError(
            f"{b_name} contains rows outside its paired A training split."
        )

    # Validation = rows excluded from the paired A training split.
    valid_idx = np.setdiff1d(
        all_indices,
        a_fit_idx,
        assume_unique=False
    )

    if len(np.intersect1d(b_fit_idx, valid_idx)) != 0:
        raise RuntimeError(
            f"Training/validation overlap detected for {b_name}."
        )

    b_target = y_array[b_fit_idx]

    n_positive = int(np.sum(b_target == 1))
    n_negative = int(np.sum(b_target == 0))

    a_positive = int(
        np.sum(y_array[a_fit_idx] == 1)
    )

    if n_positive != a_positive:
        raise RuntimeError(
            f"{b_name} does not retain all defaults from paired A."
        )

    if n_negative != 3 * n_positive:
        raise RuntimeError(
            f"{b_name} does not have the 3:1 ratio."
        )

    b_specs.append({

        "model_name": b_name,

        "b_number": b_idx,

        "paired_A_model": a_name,

        "fit_idx": b_fit_idx,

        "a_fit_idx": a_fit_idx,

        "valid_idx": valid_idx,

        "positive_count": n_positive,

        "negative_count": n_negative,

        "seed": BASE_SEED + SEED_OFFSET_B + b_idx

    })

    print(
        f"{b_name}: "
        f"train={len(b_fit_idx):,} | "
        f"validation={len(valid_idx):,} | "
        f"default={n_positive:,} | "
        f"non-default={n_negative:,}"
    )


if len(b_specs) != 10:
    raise RuntimeError("Exactly 10 B models are required.")


# Validate repeated 5-fold coverage.
validation_coverage = np.zeros(
    N_TRAIN,
    dtype=np.int8
)

for spec in b_specs:
    validation_coverage[spec["valid_idx"]] += 1

unique_coverage = np.unique(validation_coverage)

print("\nValidation coverage values:", unique_coverage)

if not np.array_equal(unique_coverage, np.array([2])):
    raise RuntimeError(
        "Expected every training row to appear in validation "
        "exactly twice (two repetitions of 5-fold CV). "
        f"Observed coverage: {unique_coverage}"
    )


# ============================================================
# 10. FEATURE ENGINEERING FUNCTIONS
# ============================================================

def fit_region_income_reference(
    raw_train,
    reference_indices
):

    reference = raw_train.iloc[
        reference_indices
    ]

    income = pd.to_numeric(
        reference["AMT_INCOME_TOTAL"],
        errors="coerce"
    )

    region = pd.to_numeric(
        reference["REGION_RATING_CLIENT"],
        errors="coerce"
    )

    reference_df = pd.DataFrame({

        "REGION_RATING_CLIENT": region.to_numpy(),

        "AMT_INCOME_TOTAL": income.to_numpy()

    })

    global_mean = float(
        reference_df["AMT_INCOME_TOTAL"].mean()
    )

    if not np.isfinite(global_mean):
        raise ValueError(
            "Invalid global income mean."
        )

    regional_means = (
        reference_df
        .dropna(
            subset=[
                "REGION_RATING_CLIENT",
                "AMT_INCOME_TOTAL"
            ]
        )
        .groupby(
            "REGION_RATING_CLIENT"
        )["AMT_INCOME_TOTAL"]
        .mean()
    )

    return regional_means, global_mean


def build_six_features(
    raw_df,
    regional_means,
    global_mean
):

    result = pd.DataFrame(
        index=np.arange(len(raw_df))
    )

    # --------------------------------------------------------
    # 1. ID_PUBLISH_AGE
    # --------------------------------------------------------

    result["ID_PUBLISH_AGE"] = (
        -pd.to_numeric(
            raw_df["DAYS_ID_PUBLISH"],
            errors="coerce"
        ).to_numpy()
    )


    # --------------------------------------------------------
    # 2. AGE_YEARS
    # --------------------------------------------------------

    result["AGE_YEARS"] = (
        -pd.to_numeric(
            raw_df["DAYS_BIRTH"],
            errors="coerce"
        ).to_numpy()
        / 365.25
    )


    # --------------------------------------------------------
    # 3 + 4. REGIONAL INCOME FEATURES
    # Region key = REGION_RATING_CLIENT
    # --------------------------------------------------------

    income = pd.to_numeric(
        raw_df["AMT_INCOME_TOTAL"],
        errors="coerce"
    )

    region = pd.to_numeric(
        raw_df["REGION_RATING_CLIENT"],
        errors="coerce"
    )

    regional_mean = (
        region
        .map(regional_means)
        .fillna(global_mean)
    )

    result["INCOME_TO_REGION_MEAN"] = (
        income / regional_mean.replace(0, np.nan)
    ).to_numpy()

    result["INCOME_GAP_FROM_REGION_MEAN"] = (
        income - regional_mean
    ).to_numpy()


    # --------------------------------------------------------
    # 5. WORKING_CASH_INDICATOR
    #
    # AMT_GOODS_PRICE = 450000
    # NAME_INCOME_TYPE = Working
    # NAME_CONTRACT_TYPE = Cash loans
    # --------------------------------------------------------

    goods_price = pd.to_numeric(
        raw_df["AMT_GOODS_PRICE"],
        errors="coerce"
    )

    income_type = (
        raw_df["NAME_INCOME_TYPE"]
        .astype(str)
    )

    contract_type = (
        raw_df["NAME_CONTRACT_TYPE"]
        .astype(str)
    )

    working_cash_mask = (

        goods_price.eq(450000)

        & income_type.eq("Working")

        & contract_type.eq("Cash loans")

    )

    result["WORKING_CASH_INDICATOR"] = (
        working_cash_mask.astype(float).to_numpy()
    )


    # --------------------------------------------------------
    # 6. WORKING_CASH_EMPLOYMENT_YEARS
    # --------------------------------------------------------

    days_employed = pd.to_numeric(
        raw_df["DAYS_EMPLOYED"],
        errors="coerce"
    )

    employment_years = (
        -days_employed / 365.25
    ).where(
        days_employed.lt(0)
    )

    working_cash_years = pd.Series(
        0.0,
        index=np.arange(len(raw_df)),
        dtype=float
    )

    working_cash_years.loc[
        working_cash_mask.to_numpy()
    ] = employment_years.to_numpy()[
        working_cash_mask.to_numpy()
    ]

    result["WORKING_CASH_EMPLOYMENT_YEARS"] = (
        working_cash_years.to_numpy()
    )


    # --------------------------------------------------------
    # CLEAN INVALID VALUES
    # --------------------------------------------------------

    result = result.replace(
        [np.inf, -np.inf],
        np.nan
    )

    result = result[NEW_FEATURES]

    return result.reset_index(drop=True)


# ============================================================
# 11. CACHE FOLD-SPECIFIC FEATURE REFERENCES
# ============================================================

print("\n" + "=" * 85)
print("PREPARING FOLD-SPECIFIC ENGINEERED FEATURES")
print("=" * 85)

fold_cache = {}

for spec in b_specs:

    model_name = spec["model_name"]

    # Regional means must be calculated from the FULL
    # corresponding A training split, before negative downsampling.
    regional_means, global_mean = (
        fit_region_income_reference(
            train_raw_df.drop(
                columns=[TARGET_COL, ID_COL]
            ),
            spec["a_fit_idx"]
        )
    )

    raw_train_features = train_raw_df.drop(
        columns=[TARGET_COL, ID_COL]
    )

    raw_test_features = test_raw_df.drop(
        columns=[ID_COL]
    )

    train_derived = build_six_features(
        raw_train_features,
        regional_means,
        global_mean
    )

    test_derived = build_six_features(
        raw_test_features,
        regional_means,
        global_mean
    )

    fold_cache[model_name] = {

        "train_features": train_derived,

        "test_features": test_derived,

        "global_income_mean": global_mean,

        "regional_group_count": len(regional_means)

    }

    print(
        model_name,
        "| regional groups:",
        len(regional_means),
        "| global mean:",
        round(global_mean, 4)
    )


# ============================================================
# 12. GENERATE ALL 64 FEATURE COMBINATIONS
# ============================================================

combination_specs = [

    {
        "combination_id": "C00_BASELINE",
        "features": tuple()
    }

]

combination_number = 0

for subset_size in range(1, len(NEW_FEATURES) + 1):

    for subset in combinations(
        NEW_FEATURES,
        subset_size
    ):

        combination_number += 1

        combination_specs.append({

            "combination_id": f"C{combination_number:02d}",

            "features": tuple(subset)

        })


if len(combination_specs) != EXPECTED_COMBINATIONS:
    raise RuntimeError(
        f"Expected 64 combinations, got {len(combination_specs)}."
    )


print("\n" + "=" * 85)
print("FEATURE COMBINATION PLAN")
print("=" * 85)

print("Optional features:", len(NEW_FEATURES))
print("Non-empty combinations:", 2 ** len(NEW_FEATURES) - 1)
print("Baseline combinations:", 1)
print("Total combinations:", len(combination_specs))
print("B models per combination:", B_MODEL_COUNT)
print(
    "Total models:",
    len(combination_specs) * B_MODEL_COUNT
)


combination_plan_rows = []

model_order_rows = []

global_model_index = 0

for combo_idx, combo_spec in enumerate(combination_specs):

    combo_id = combo_spec["combination_id"]
    combo_features = combo_spec["features"]

    combination_plan_rows.append({

        "combination_index": combo_idx,

        "combination_id": combo_id,

        "feature_count": len(combo_features),

        "optional_features": (
            "; ".join(combo_features)
            if combo_features
            else "NONE"
        ),

        "fixed_feature": "CREDIT_TERM"

    })

    for spec in b_specs:

        global_model_index += 1

        model_order_rows.append({

            "matrix_row": global_model_index - 1,

            "combination_index": combo_idx,

            "combination_id": combo_id,

            "model_name": spec["model_name"],

            "paired_A_model": spec["paired_A_model"],

            "b_number": spec["b_number"],

            "optional_features": (
                "; ".join(combo_features)
                if combo_features
                else "NONE"
            )

        })


combination_plan_df = pd.DataFrame(
    combination_plan_rows
)

model_order_df = pd.DataFrame(
    model_order_rows
)

combination_plan_df.to_csv(
    OUTPUT_DIR / "combination_plan.csv",
    index=False
)

model_order_df.to_csv(
    OUTPUT_DIR / "model_order.csv",
    index=False
)


if len(model_order_df) != EXPECTED_TOTAL_MODELS:
    raise RuntimeError(
        "Model order table does not contain 640 models."
    )


# ============================================================
# 13. CATBOOST PARAMETERS
# ============================================================

CATBOOST_PARAMS = {

    "iterations": MAX_ITERATIONS,

    "depth": DEPTH,

    "learning_rate": LEARNING_RATE,

    "l2_leaf_reg": L2_LEAF_REG,

    "loss_function": "Logloss",

    "eval_metric": "AUC",

    "grow_policy": GROW_POLICY,

    "allow_writing_files": False,

    "thread_count": -1,

    "verbose": LOG_PERIOD

}


if TASK_TYPE.upper() == "GPU":

    CATBOOST_PARAMS["task_type"] = "GPU"
    CATBOOST_PARAMS["devices"] = "0"

elif TASK_TYPE.upper() != "CPU":

    raise ValueError(
        "TASK_TYPE must be CPU or GPU."
    )


print("\nCatBoost parameters:")
print(CATBOOST_PARAMS)


# ============================================================
# 14. METRIC AND FEATURE MATRIX UTILITIES
# ============================================================

def calculate_metrics(y_true, probabilities):

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )

    probabilities_clipped = np.clip(
        probabilities,
        1e-7,
        1 - 1e-7
    )

    y_true = np.asarray(
        y_true,
        dtype=np.int8
    )

    return {

        "roc_auc": float(
            roc_auc_score(
                y_true,
                probabilities
            )
        ),

        "average_precision": float(
            average_precision_score(
                y_true,
                probabilities
            )
        ),

        "logloss": float(
            log_loss(
                y_true,
                probabilities_clipped,
                labels=[0, 1]
            )
        ),

        "brier_score": float(
            brier_score_loss(
                y_true,
                probabilities
            )
        )

    }


def build_model_matrix(
    base_matrix,
    engineered_matrix,
    indices,
    selected_features
):

    base_part = (
        base_matrix
        .iloc[indices]
        .reset_index(drop=True)
    )

    if not selected_features:
        return base_part

    engineered_part = (
        engineered_matrix
        .iloc[indices][list(selected_features)]
        .reset_index(drop=True)
    )

    result = pd.concat(
        [base_part, engineered_part],
        axis=1
    )

    if not result.columns.is_unique:
        raise RuntimeError(
            "Duplicate columns found in model matrix."
        )

    return result


def build_test_matrix(
    selected_features,
    engineered_test
):

    base_part = X_base_test.copy().reset_index(drop=True)

    if not selected_features:
        return base_part

    engineered_part = (
        engineered_test[list(selected_features)]
        .reset_index(drop=True)
    )

    result = pd.concat(
        [base_part, engineered_part],
        axis=1
    )

    if not result.columns.is_unique:
        raise RuntimeError(
            "Duplicate test columns found."
        )

    return result


def sha256_array(values):

    values = np.asarray(
        values,
        dtype=np.int64
    )

    return hashlib.sha256(
        values.tobytes()
    ).hexdigest()


def save_json(path, data):

    temp_path = path.with_suffix(
        path.suffix + ".tmp"
    )

    with open(
        temp_path,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            data,
            f,
            indent=2,
            ensure_ascii=False
        )

    os.replace(
        temp_path,
        path
    )


# ============================================================
# 15. ROOT VECTOR EXTRACTION WITH VARIABLE TREE COUNT
# ============================================================

NUMERICAL_TYPES = {
    "FloatFeature"
}

CATEGORICAL_TYPES = {
    "OneHotFeature",
    "OnlineCtr"
}


def extract_root_vector(json_path):

    with open(
        json_path,
        "r",
        encoding="utf-8"
    ) as f:

        model_json = json.load(f)

    trees = model_json.get("oblivious_trees")

    if trees is None:
        raise RuntimeError(
            f"Missing oblivious_trees in {json_path}"
        )

    features_info = model_json.get(
        "features_info",
        {}
    )

    float_lookup = {}

    for item in features_info.get(
        "float_features",
        []
    ):

        feature_index = item.get(
            "feature_index"
        )

        flat_index = item.get(
            "flat_feature_index"
        )

        feature_id = item.get(
            "feature_id"
        )

        if feature_index is not None:

            if feature_id:
                float_lookup[int(feature_index)] = str(feature_id)

            elif flat_index is not None:
                flat_index = int(flat_index)

                if flat_index < len(BASE_FEATURES) + len(NEW_FEATURES):
                    float_lookup[int(feature_index)] = (
                        ALL_MODEL_FEATURES[flat_index]
                    )


    cat_lookup = {}

    for item in features_info.get(
        "categorical_features",
        []
    ):

        feature_index = item.get(
            "feature_index"
        )

        flat_index = item.get(
            "flat_feature_index"
        )

        feature_id = item.get(
            "feature_id"
        )

        if feature_index is not None:

            if feature_id:
                cat_lookup[int(feature_index)] = str(feature_id)

            elif flat_index is not None:

                flat_index = int(flat_index)

                if flat_index < len(ALL_MODEL_FEATURES):
                    cat_lookup[int(feature_index)] = (
                        ALL_MODEL_FEATURES[flat_index]
                    )


    vector = np.full(
        len(trees),
        2,
        dtype=np.uint8
    )

    audit_rows = []

    for tree_idx, tree in enumerate(trees):

        splits = tree.get("splits") or []

        if len(splits) == 0:

            vector[tree_idx] = 2

            audit_rows.append({

                "tree_index": tree_idx + 1,

                "root_value": 2,

                "root_kind": "no_split",

                "root_split_type": None,

                "root_feature_name": None,

                "root_feature_index": None,

                "root_split_details": None

            })

            continue

        # SymmetricTree root is the last split in JSON order.
        root_split = splits[-1]

        split_type = root_split.get("split_type")

        if split_type in NUMERICAL_TYPES:

            value = 0
            kind = "numerical"

            feature_index = root_split.get(
                "float_feature_index"
            )

            feature_name = (
                float_lookup.get(int(feature_index))
                if feature_index is not None
                else None
            )

        elif split_type in CATEGORICAL_TYPES:

            value = 1
            kind = "categorical"

            feature_index = root_split.get(
                "cat_feature_index"
            )

            if feature_index is not None:

                feature_name = cat_lookup.get(
                    int(feature_index)
                )

            else:

                # OnlineCtr may represent a categorical combination.
                feature_name = "CTR"

                feature_index = root_split.get(
                    "split_index"
                )

        else:

            raise RuntimeError(
                f"Unknown root split type: {split_type} "
                f"in {json_path}, tree {tree_idx + 1}"
            )

        vector[tree_idx] = value

        audit_rows.append({

            "tree_index": tree_idx + 1,

            "root_value": value,

            "root_kind": kind,

            "root_split_type": split_type,

            "root_feature_name": feature_name,

            "root_feature_index": feature_index,

            "root_split_details": json.dumps(
                root_split,
                ensure_ascii=False
            )

        })


    if len(vector) != len(trees):
        raise RuntimeError(
            "Root vector/tree count mismatch."
        )

    if not np.isin(vector, [0, 1, 2]).all():
        raise RuntimeError(
            "Invalid root-vector encoding."
        )

    return vector, audit_rows


# Complete feature list for JSON flat-index mapping.
ALL_MODEL_FEATURES = (
    BASE_FEATURES + NEW_FEATURES
)


# ============================================================
# 16. SAVE EXPERIMENT CONFIGURATION
# ============================================================

FEATURE_COMBINATIONS = [
    {
        "combination_id": item["combination_id"],
        "features": list(item["features"])
    }
    for item in combination_specs
]


run_metadata = {

    "experiment": "B_PLUS_ALL_6_FEATURE_COMBINATIONS",

    "base_experiment": str(
        BASE_EXPERIMENT_DIR
    ),

    "total_combinations": EXPECTED_COMBINATIONS,

    "models_per_combination": B_MODEL_COUNT,

    "total_models": EXPECTED_TOTAL_MODELS,

    "fixed_feature": "CREDIT_TERM",

    "new_optional_features": NEW_FEATURES,

    "tree_configuration": {

        "maximum_iterations": MAX_ITERATIONS,

        "early_stopping_rounds": EARLY_STOPPING_ROUNDS,

        "use_best_model": True,

        "eval_metric": "AUC"

    },

    "catboost_parameters": CATBOOST_PARAMS,

    "sampling": {

        "source": "Original FULLPOS_3TO1 training_sample_manifest.csv",

        "negative_to_positive_ratio": "3:1",

        "all_defaults_retained": True,

        "same_ten_samples_for_every_combination": True,

        "validation": (
            "Complement of paired A model training indices; "
            "two repetitions of five-fold validation."
        )

    },

    "region_income_reference": {

        "grouping_feature": "REGION_RATING_CLIENT",

        "calculated_from": (
            "Full corresponding A training split, "
            "before negative downsampling"
        ),

        "target_used": False,

        "unseen_region_fallback": "Global mean"

    },

    "root_vector_encoding": {

        "0": "Numerical",

        "1": "Categorical",

        "2": "No split",

        "255": "Padding only"

    },

    "combinations": FEATURE_COMBINATIONS,

    "test_rows": N_TEST,

    "training_rows": N_TRAIN

}


save_json(
    OUTPUT_DIR / "experiment_metadata.json",
    run_metadata
)


# ============================================================
# 17. RESUMABLE TRAINING OF 640 MODELS
# ============================================================

print("\n" + "=" * 85)
print("STARTING 640-MODEL FEATURE SEARCH")
print("=" * 85)

model_metrics_rows = []

root_summary_rows = []

completed_count = 0

failed_count = 0

total_started = perf_counter()


for combo_spec in combination_specs:

    combo_id = combo_spec["combination_id"]

    selected_features = combo_spec["features"]

    combo_model_dir = MODEL_DIR / combo_id

    combo_vector_dir = (
        VECTOR_INDIVIDUAL_DIR / combo_id
    )

    combo_audit_dir = (
        AUDIT_DIR / combo_id
    )

    combo_prediction_dir = (
        PREDICTION_CHECKPOINT_DIR / combo_id
    )

    for directory in [
        combo_model_dir,
        combo_vector_dir,
        combo_audit_dir,
        combo_prediction_dir
    ]:
        directory.mkdir(
            parents=True,
            exist_ok=True
        )


    print("\n" + "#" * 85)

    print(
        f"COMBINATION: {combo_id}"
    )

    print(
        "Optional features:",
        list(selected_features)
        if selected_features
        else "NONE - BASELINE"
    )

    print(
        "Total input features:",
        len(BASE_FEATURES) + len(selected_features)
    )

    print("#" * 85)


    # --------------------------------------------------------
    # Train the same ten B models for this combination.
    # --------------------------------------------------------

    for spec in b_specs:

        model_name = spec["model_name"]

        b_idx = spec["b_number"]

        fit_idx = spec["fit_idx"]

        valid_idx = spec["valid_idx"]

        fold_features = fold_cache[model_name]

        # ----------------------------------------------------
        # Individual artifact paths
        # ----------------------------------------------------

        model_path = (
            combo_model_dir /
            f"{model_name}.cbm"
        )

        json_path = (
            combo_model_dir /
            f"{model_name}.json"
        )

        vector_path = (
            combo_vector_dir /
            f"{model_name}_root.npy"
        )

        vector_csv_path = (
            combo_vector_dir /
            f"{model_name}_root.csv"
        )

        audit_path = (
            combo_audit_dir /
            f"{model_name}_root_audit.csv"
        )

        prediction_path = (
            combo_prediction_dir /
            f"{model_name}_predictions.npz"
        )

        info_path = (
            combo_model_dir /
            f"{model_name}_metadata.json"
        )


        # ----------------------------------------------------
        # Stable run signature
        # ----------------------------------------------------

        signature_payload = {

            "combination_id": combo_id,

            "features": list(selected_features),

            "model_name": model_name,

            "fit_indices_hash": sha256_array(fit_idx),

            "validation_indices_hash": sha256_array(valid_idx),

            "seed": spec["seed"],

            "parameters": CATBOOST_PARAMS,

            "early_stopping_rounds": EARLY_STOPPING_ROUNDS

        }

        run_signature = hashlib.sha256(
            json.dumps(
                signature_payload,
                sort_keys=True
            ).encode("utf-8")
        ).hexdigest()


        # ----------------------------------------------------
        # RESUME: skip an already completed model
        # ----------------------------------------------------

        completed_files_exist = all(
            p.exists()
            for p in [
                model_path,
                json_path,
                vector_path,
                vector_csv_path,
                audit_path,
                prediction_path,
                info_path
            ]
        )

        if completed_files_exist:

            try:

                with open(
                    info_path,
                    "r",
                    encoding="utf-8"
                ) as f:

                    saved_info = json.load(f)

                if saved_info.get("run_signature") == run_signature:

                    with np.load(
                        prediction_path,
                        allow_pickle=False
                    ) as saved_pred:

                        saved_valid_idx = saved_pred[
                            "valid_indices"
                        ]

                        saved_valid_prob = saved_pred[
                            "valid_probabilities"
                        ]

                        saved_test_prob = saved_pred[
                            "test_probabilities"
                        ]

                        saved_metrics = json.loads(
                            str(
                                saved_pred["metrics_json"].item()
                            )
                        )

                    if (
                        np.array_equal(
                            saved_valid_idx,
                            valid_idx
                        )
                        and len(saved_valid_prob) == len(valid_idx)
                        and len(saved_test_prob) == N_TEST
                        and np.isfinite(saved_test_prob).all()
                        and np.isfinite(saved_valid_prob).all()
                    ):

                        completed_count += 1

                        model_metrics_rows.append({

                            "combination_id": combo_id,

                            "model_name": model_name,

                            "b_number": b_idx,

                            "validation_auc": saved_metrics["roc_auc"],

                            "validation_ap": saved_metrics["average_precision"],

                            "validation_logloss": saved_metrics["logloss"],

                            "validation_brier": saved_metrics["brier_score"],

                            "best_iteration": saved_info["best_iteration"],

                            "tree_count": saved_info["tree_count"],

                            "runtime_seconds": saved_info["runtime_seconds"],

                            "status": "RESTORED"

                        })

                        print(
                            f"[RESTORED] {combo_id} / {model_name} "
                            f"| trees={saved_info['tree_count']} "
                            f"| AUC={saved_metrics['roc_auc']:.6f}"
                        )

                        continue

            except Exception as exc:

                print(
                    "Checkpoint invalid; retraining:",
                    combo_id,
                    model_name,
                    str(exc)
                )


        # ----------------------------------------------------
        # Prepare training, validation and test matrices.
        # ----------------------------------------------------

        X_train_run = build_model_matrix(

            X_base_train,

            fold_features["train_features"],

            fit_idx,

            selected_features

        )

        X_valid_run = build_model_matrix(

            X_base_train,

            fold_features["train_features"],

            valid_idx,

            selected_features

        )

        X_test_run = build_test_matrix(

            selected_features,

            fold_features["test_features"]

        )

        y_train_run = y_array[fit_idx]

        y_valid_run = y_array[valid_idx]


        if len(X_train_run) != len(fit_idx):
            raise RuntimeError(
                f"Training row mismatch: {model_name}"
            )

        if len(X_valid_run) != len(valid_idx):
            raise RuntimeError(
                f"Validation row mismatch: {model_name}"
            )

        if list(X_train_run.columns) != list(X_valid_run.columns):
            raise RuntimeError(
                f"Training/validation feature order mismatch: {model_name}"
            )

        if list(X_train_run.columns) != list(X_test_run.columns):
            raise RuntimeError(
                f"Training/test feature order mismatch: {model_name}"
            )


        # ----------------------------------------------------
        # Train model using early stopping.
        # ----------------------------------------------------

        print("\n" + "-" * 75)

        print(
            f"TRAINING: {combo_id} / {model_name}"
        )

        print("Training rows:", len(fit_idx))

        print("Validation rows:", len(valid_idx))

        print("Features:", X_train_run.shape[1])

        print("Seed:", spec["seed"])


        model = CatBoostClassifier(

            **CATBOOST_PARAMS,

            random_seed=int(spec["seed"])

        )

        started = perf_counter()

        try:

            model.fit(

                X_train_run,

                y_train_run,

                cat_features=CAT_FEATURES,

                eval_set=(
                    X_valid_run,
                    y_valid_run
                ),

                early_stopping_rounds=EARLY_STOPPING_ROUNDS,

                use_best_model=True,

                verbose=LOG_PERIOD

            )


            # ------------------------------------------------
            # Determine actual tree count.
            # ------------------------------------------------

            tree_count = int(
                model.tree_count_
            )

            best_iteration = int(
                model.get_best_iteration()
            )

            if best_iteration < 0:
                best_iteration = tree_count - 1

            if tree_count <= 0:
                raise RuntimeError(
                    f"Invalid tree count in {model_name}."
                )


            # ------------------------------------------------
            # Save CatBoost model BEFORE predictions.
            # This preserves every completed model for blending.
            # ------------------------------------------------

            model.save_model(
                str(model_path),
                format="cbm"
            )

            model.save_model(
                str(json_path),
                format="json"
            )


            # ------------------------------------------------
            # Validation and test predictions.
            # ------------------------------------------------

            classes = list(
                model.classes_
            )

            positive_indices = [

                i for i, cls in enumerate(classes)

                if cls == 1 or str(cls) == "1"

            ]

            if len(positive_indices) != 1:
                raise RuntimeError(
                    f"Cannot identify TARGET=1 in {model_name}."
                )

            positive_idx = positive_indices[0]


            valid_probabilities = model.predict_proba(
                X_valid_run
            )[:, positive_idx]

            test_probabilities = model.predict_proba(
                X_test_run
            )[:, positive_idx]


            valid_probabilities = np.asarray(
                valid_probabilities,
                dtype=np.float64
            )

            test_probabilities = np.asarray(
                test_probabilities,
                dtype=np.float64
            )


            if (
                not np.isfinite(valid_probabilities).all()
                or not np.isfinite(test_probabilities).all()
            ):
                raise RuntimeError(
                    f"Non-finite predictions in {model_name}."
                )


            # ------------------------------------------------
            # Validation metrics.
            # ------------------------------------------------

            metrics = calculate_metrics(
                y_valid_run,
                valid_probabilities
            )


            # ------------------------------------------------
            # Extract variable-length root vector.
            # ------------------------------------------------

            vector, audit_rows = extract_root_vector(
                json_path
            )

            if len(vector) != tree_count:
                raise RuntimeError(
                    f"Root vector length {len(vector)} "
                    f"does not match tree count {tree_count}."
                )


            for row in audit_rows:

                row["combination_id"] = combo_id

                row["model_name"] = model_name


            audit_df = pd.DataFrame(
                audit_rows
            )


            # ------------------------------------------------
            # Save root vector and detailed audit.
            # ------------------------------------------------

            np.save(
                vector_path,
                vector
            )

            vector_columns = [
                f"TREE_{i:04d}"
                for i in range(1, len(vector) + 1)
            ]

            pd.DataFrame(
                [vector],
                columns=vector_columns
            ).to_csv(
                vector_csv_path,
                index=False
            )

            audit_df.to_csv(
                audit_path,
                index=False
            )


            # ------------------------------------------------
            # Root summary.
            # ------------------------------------------------

            n_numerical = int(
                np.sum(vector == 0)
            )

            n_categorical = int(
                np.sum(vector == 1)
            )

            n_no_split = int(
                np.sum(vector == 2)
            )

            root_feature_counts = {}

            for feature in NEW_FEATURES:

                root_feature_counts[
                    f"root_{feature}_count"
                ] = int(
                    (
                        audit_df["root_feature_name"]
                        == feature
                    ).sum()
                )

            root_summary = {

                "combination_id": combo_id,

                "model_name": model_name,

                "tree_count": tree_count,

                "best_iteration": best_iteration,

                "numerical_roots": n_numerical,

                "categorical_roots": n_categorical,

                "no_split_trees": n_no_split,

                "numerical_fraction": n_numerical / tree_count,

                "categorical_fraction": n_categorical / tree_count,

                "no_split_fraction": n_no_split / tree_count,

                "new_feature_root_count": sum(
                    root_feature_counts.values()
                )

            }

            root_summary.update(
                root_feature_counts
            )

            root_summary_rows.append(
                root_summary
            )


            # ------------------------------------------------
            # Save all predictions as resumable checkpoint.
            # ------------------------------------------------

            metrics_json = json.dumps(
                metrics
            )

            temp_prediction_path = (
                prediction_path.with_suffix(".npz.tmp")
            )

            with open(
                temp_prediction_path,
                "wb"
            ) as f:

                np.savez_compressed(

                    f,

                    run_signature=np.asarray(run_signature),

                    valid_indices=valid_idx.astype(np.int64),

                    valid_probabilities=valid_probabilities.astype(np.float32),

                    test_probabilities=test_probabilities.astype(np.float32),

                    metrics_json=np.asarray(metrics_json)

                )

            os.replace(
                temp_prediction_path,
                prediction_path
            )


            # ------------------------------------------------
            # Save model metadata.
            # ------------------------------------------------

            runtime_seconds = (
                perf_counter() - started
            )

            model_info = {

                "run_signature": run_signature,

                "combination_id": combo_id,

                "optional_features": list(selected_features),

                "fixed_feature": "CREDIT_TERM",

                "model_name": model_name,

                "paired_A_model": spec["paired_A_model"],

                "training_rows": len(fit_idx),

                "validation_rows": len(valid_idx),

                "default_count": spec["positive_count"],

                "non_default_count": spec["negative_count"],

                "negative_to_positive_ratio": "3:1",

                "random_seed": spec["seed"],

                "tree_count": tree_count,

                "best_iteration": best_iteration,

                "maximum_iterations": MAX_ITERATIONS,

                "early_stopping_rounds": EARLY_STOPPING_ROUNDS,

                "validation_metrics": metrics,

                "runtime_seconds": round(runtime_seconds, 2),

                "model_file": str(model_path),

                "json_file": str(json_path),

                "root_vector_file": str(vector_path),

                "root_audit_file": str(audit_path),

                "prediction_checkpoint": str(prediction_path)

            }

            save_json(
                info_path,
                model_info
            )


            # ------------------------------------------------
            # Record completion.
            # ------------------------------------------------

            completed_count += 1

            model_metrics_rows.append({

                "combination_id": combo_id,

                "model_name": model_name,

                "b_number": b_idx,

                "validation_auc": metrics["roc_auc"],

                "validation_ap": metrics["average_precision"],

                "validation_logloss": metrics["logloss"],

                "validation_brier": metrics["brier_score"],

                "best_iteration": best_iteration,

                "tree_count": tree_count,

                "runtime_seconds": round(runtime_seconds, 2),

                "status": "OK"

            })


            # Save incremental metrics after every model.
            pd.DataFrame(
                model_metrics_rows
            ).to_csv(
                OUTPUT_DIR / "per_model_metrics_progress.csv",
                index=False
            )


            print("\nMODEL COMPLETED")

            print("Tree count:", tree_count)

            print("Best iteration:", best_iteration)

            print(
                "Validation ROC-AUC:",
                f"{metrics['roc_auc']:.6f}"
            )

            print(
                "Validation AP:",
                f"{metrics['average_precision']:.6f}"
            )

            print(
                "Numerical roots:",
                n_numerical
            )

            print(
                "Categorical roots:",
                n_categorical
            )

            print(
                "New feature roots:",
                root_summary["new_feature_root_count"]
            )

            print(
                "Saved model:",
                model_path.name
            )


        except Exception as exc:

            failed_count += 1

            print(
                "\nMODEL FAILED:",
                combo_id,
                model_name
            )

            print(
                "Error:",
                repr(exc)
            )

            raise


        finally:

            if "model" in locals():
                del model

            if "X_train_run" in locals():
                del X_train_run

            if "X_valid_run" in locals():
                del X_valid_run

            if "X_test_run" in locals():
                del X_test_run

            gc.collect()


print("\n" + "=" * 85)
print("ALL MODEL TRAINING LOOPS COMPLETED")
print("=" * 85)

print("Models completed/restored:", completed_count)

print("Models failed:", failed_count)

print(
    "Training runtime:",
    round(perf_counter() - total_started, 2),
    "seconds"
)


# ============================================================
# 18. FINALIZE MODEL ORDER AND TEST PREDICTION MATRIX
# ============================================================

print("\nBuilding blend-ready test prediction matrix...")


MODEL_COUNT = EXPECTED_TOTAL_MODELS

test_prediction_matrix = np.zeros(
    (MODEL_COUNT, N_TEST),
    dtype=np.float32
)

oof_sum = np.zeros(
    (EXPECTED_COMBINATIONS, N_TRAIN),
    dtype=np.float64
)

oof_count = np.zeros(
    (EXPECTED_COMBINATIONS, N_TRAIN),
    dtype=np.uint8
)

combo_test_prediction_matrix = np.zeros(
    (EXPECTED_COMBINATIONS, N_TEST),
    dtype=np.float32
)

per_model_final_rows = []

root_summary_final_rows = []

root_vectors_list = []

root_vector_lengths = []

root_audit_files = []


for _, order_row in model_order_df.iterrows():

    matrix_row = int(
        order_row["matrix_row"]
    )

    combo_idx = int(
        order_row["combination_index"]
    )

    combo_id = order_row["combination_id"]

    model_name = order_row["model_name"]


    prediction_path = (
        PREDICTION_CHECKPOINT_DIR /
        combo_id /
        f"{model_name}_predictions.npz"
    )

    model_info_path = (
        MODEL_DIR /
        combo_id /
        f"{model_name}_metadata.json"
    )

    vector_path = (
        VECTOR_INDIVIDUAL_DIR /
        combo_id /
        f"{model_name}_root.npy"
    )

    audit_path = (
        AUDIT_DIR /
        combo_id /
        f"{model_name}_root_audit.csv"
    )


    required_files = [
        prediction_path,
        model_info_path,
        vector_path,
        audit_path,
        MODEL_DIR / combo_id / f"{model_name}.cbm",
        MODEL_DIR / combo_id / f"{model_name}.json"
    ]

    if not all(
        p.exists()
        for p in required_files
    ):
        raise RuntimeError(
            f"Incomplete model artifacts: {combo_id}/{model_name}"
        )


    with np.load(
        prediction_path,
        allow_pickle=False
    ) as pred:

        valid_idx = pred[
            "valid_indices"
        ].astype(np.int64)

        valid_prob = pred[
            "valid_probabilities"
        ].astype(np.float64)

        test_prob = pred[
            "test_probabilities"
        ].astype(np.float32)

        metrics = json.loads(
            str(pred["metrics_json"].item())
        )


    if len(test_prob) != N_TEST:
        raise RuntimeError(
            f"Invalid test prediction length: {model_name}"
        )

    if len(valid_prob) != len(valid_idx):
        raise RuntimeError(
            f"Invalid validation prediction length: {model_name}"
        )

    if not np.isfinite(test_prob).all():
        raise RuntimeError(
            f"Non-finite test predictions: {model_name}"
        )


    test_prediction_matrix[matrix_row] = test_prob

    oof_sum[combo_idx, valid_idx] += valid_prob

    oof_count[combo_idx, valid_idx] += 1


    with open(
        model_info_path,
        "r",
        encoding="utf-8"
    ) as f:

        info = json.load(f)


    per_model_final_rows.append({

        "combination_id": combo_id,

        "model_name": model_name,

        "b_number": int(order_row["b_number"]),

        "validation_auc": metrics["roc_auc"],

        "validation_ap": metrics["average_precision"],

        "validation_logloss": metrics["logloss"],

        "validation_brier": metrics["brier_score"],

        "best_iteration": info["best_iteration"],

        "tree_count": info["tree_count"],

        "runtime_seconds": info["runtime_seconds"],

        "status": "OK"

    })


    # Variable-length root vectors.
    vector = np.load(
        vector_path
    ).astype(np.uint8)

    if len(vector) != info["tree_count"]:
        raise RuntimeError(
            f"Root vector length mismatch: {model_name}"
        )

    root_vectors_list.append(vector)

    root_vector_lengths.append(
        len(vector)
    )

    root_audit_files.append(
        audit_path
    )

    # Root summary.
    n_num = int(np.sum(vector == 0))
    n_cat = int(np.sum(vector == 1))
    n_no = int(np.sum(vector == 2))

    audit_small = pd.read_csv(
        audit_path,
        usecols=["root_feature_name"]
    )

    feature_counts = {}

    for feature in NEW_FEATURES:

        feature_counts[
            f"root_{feature}_count"
        ] = int(
            audit_small["root_feature_name"]
            .eq(feature)
            .sum()
        )

    row = {

        "combination_id": combo_id,

        "model_name": model_name,

        "tree_count": len(vector),

        "numerical_roots": n_num,

        "categorical_roots": n_cat,

        "no_split_trees": n_no,

        "numerical_fraction": n_num / len(vector),

        "categorical_fraction": n_cat / len(vector),

        "no_split_fraction": n_no / len(vector)

    }

    row.update(feature_counts)

    root_summary_final_rows.append(row)


# ============================================================
# 19. VALIDATE OOF COVERAGE
# ============================================================

print("\nChecking OOF coverage...")

for combo_idx, combo_spec in enumerate(combination_specs):

    coverage = oof_count[combo_idx]

    if not np.all(coverage == 2):

        raise RuntimeError(
            f"OOF coverage error in {combo_spec['combination_id']}. "
            f"Expected 2 predictions per training row; "
            f"observed counts: {np.unique(coverage)}"
        )


oof_predictions_mean = (
    oof_sum / oof_count
).astype(np.float32)


if not np.isfinite(oof_predictions_mean).all():
    raise RuntimeError(
        "Invalid OOF prediction matrix."
    )


# ============================================================
# 20. SAVE BLEND-READY PREDICTION MATRICES
# ============================================================

np.save(
    OUTPUT_DIR / "test_predictions.npy",
    test_prediction_matrix
)

np.save(
    OUTPUT_DIR / "oof_predictions_mean.npy",
    oof_predictions_mean
)

np.save(
    OUTPUT_DIR / "oof_coverage.npy",
    oof_count
)

np.save(
    OUTPUT_DIR / "test_ids.npy",
    test_ids
)

print("\nSaved all-model test prediction matrix:")
print(test_prediction_matrix.shape)

print("Saved OOF prediction matrix:")
print(oof_predictions_mean.shape)


# ============================================================
# 21. BUILD PADDED ROOT VECTOR MATRIX
# ============================================================

print("\nBuilding variable-length root vector matrix...")

max_tree_count = max(
    root_vector_lengths
)

padded_root_vectors = np.full(
    (
        EXPECTED_TOTAL_MODELS,
        max_tree_count
    ),
    255,
    dtype=np.uint8
)

for i, vector in enumerate(root_vectors_list):

    padded_root_vectors[
        i,
        :len(vector)
    ] = vector


np.save(
    VECTOR_DIR / "root_vectors_padded.npy",
    padded_root_vectors
)

np.save(
    VECTOR_DIR / "root_vector_lengths.npy",
    np.asarray(
        root_vector_lengths,
        dtype=np.int32
    )
)

root_vector_order_df = model_order_df.copy()

root_vector_order_df["tree_count"] = (
    root_vector_lengths
)

root_vector_order_df.to_csv(
    VECTOR_DIR / "root_vector_model_order.csv",
    index=False
)


# ============================================================
# 22. SAVE ROOT VECTOR SUMMARY
# ============================================================

root_summary_df = pd.DataFrame(
    root_summary_final_rows
)

root_summary_df.to_csv(
    VECTOR_DIR / "root_vector_summary.csv",
    index=False
)


# ============================================================
# 23. MERGE ALL ROOT SPLIT AUDITS
# ============================================================

print("\nCombining root split audits...")

combined_audit_path = (
    OUTPUT_DIR / "root_split_audit_all_models.csv.gz"
)

with gzip.open(
    combined_audit_path,
    "wt",
    encoding="utf-8",
    newline=""
) as gzip_file:

    first_file = True

    for audit_path in root_audit_files:

        audit_part = pd.read_csv(
            audit_path
        )

        audit_part.to_csv(
            gzip_file,
            index=False,
            header=first_file
        )

        first_file = False

        del audit_part


# ============================================================
# 24. SAVE PER-MODEL VALIDATION METRICS
# ============================================================

per_model_metrics_df = pd.DataFrame(
    per_model_final_rows
)

per_model_metrics_df = per_model_metrics_df.sort_values(
    [
        "combination_id",
        "b_number"
    ]
)

per_model_metrics_df.to_csv(
    OUTPUT_DIR / "per_model_validation_metrics.csv",
    index=False
)


# ============================================================
# 25. COMBINATION-LEVEL OOF METRICS AND SUBMISSIONS
# ============================================================

print("\nGenerating 64 combination submissions...")

combination_summary_rows = []

for combo_idx, combo_spec in enumerate(combination_specs):

    combo_id = combo_spec["combination_id"]

    combo_features = combo_spec["features"]

    combo_models = per_model_metrics_df[
        per_model_metrics_df["combination_id"] == combo_id
    ].copy()

    if len(combo_models) != B_MODEL_COUNT:
        raise RuntimeError(
            f"{combo_id} does not have exactly 10 completed models."
        )


    # Mean OOF prediction across the two validation repetitions.
    oof_probabilities = (
        oof_predictions_mean[combo_idx]
    )

    oof_metrics = calculate_metrics(
        y_array,
        oof_probabilities
    )


    # Average fold-level metrics across 10 B models.
    mean_fold_auc = float(
        combo_models["validation_auc"].mean()
    )

    std_fold_auc = float(
        combo_models["validation_auc"].std(ddof=1)
    )

    mean_fold_ap = float(
        combo_models["validation_ap"].mean()
    )

    mean_fold_logloss = float(
        combo_models["validation_logloss"].mean()
    )


    # Mean test prediction across the ten B models.
    start_row = combo_idx * B_MODEL_COUNT

    end_row = start_row + B_MODEL_COUNT

    combo_test_probabilities = (
        test_prediction_matrix[
            start_row:end_row
        ].mean(axis=0)
    )

    combo_test_prediction_matrix[
        combo_idx
    ] = combo_test_probabilities


    # --------------------------------------------------------
    # Save one submission for every feature combination.
    # --------------------------------------------------------

    submission = pd.DataFrame({

        ID_COL: test_ids,

        TARGET_COL: combo_test_probabilities

    })

    submission_path = (
        SUBMISSION_DIR /
        f"submission_{combo_id}.csv"
    )

    submission.to_csv(
        submission_path,
        index=False
    )


    combination_summary_rows.append({

        "combination_index": combo_idx,

        "combination_id": combo_id,

        "optional_feature_count": len(combo_features),

        "optional_features": (
            "; ".join(combo_features)
            if combo_features
            else "NONE"
        ),

        "total_features": (
            len(BASE_FEATURES) + len(combo_features)
        ),

        "number_of_models": len(combo_models),

        "mean_fold_auc": mean_fold_auc,

        "std_fold_auc": std_fold_auc,

        "mean_fold_average_precision": mean_fold_ap,

        "mean_fold_logloss": mean_fold_logloss,

        "pooled_oof_auc": oof_metrics["roc_auc"],

        "pooled_oof_average_precision": (
            oof_metrics["average_precision"]
        ),

        "pooled_oof_logloss": oof_metrics["logloss"],

        "pooled_oof_brier": oof_metrics["brier_score"],

        "mean_tree_count": float(
            combo_models["tree_count"].mean()
        ),

        "median_tree_count": float(
            combo_models["tree_count"].median()
        ),

        "min_tree_count": int(
            combo_models["tree_count"].min()
        ),

        "max_tree_count": int(
            combo_models["tree_count"].max()
        ),

        "test_prediction_mean": float(
            combo_test_probabilities.mean()
        ),

        "test_prediction_std": float(
            combo_test_probabilities.std()
        ),

        "submission_file": submission_path.name

    })


# Save combination test-prediction matrix.
np.save(
    OUTPUT_DIR / "combination_test_predictions.npy",
    combo_test_prediction_matrix
)


# ============================================================
# 26. RANK ALL COMBINATIONS
# ============================================================

combination_summary_df = pd.DataFrame(
    combination_summary_rows
)

combination_summary_df = combination_summary_df.sort_values(
    [
        "pooled_oof_auc",
        "mean_fold_auc",
        "pooled_oof_average_precision",
        "pooled_oof_logloss"
    ],
    ascending=[
        False,
        False,
        False,
        True
    ]
).reset_index(drop=True)

combination_summary_df.insert(
    0,
    "rank",
    np.arange(
        1,
        len(combination_summary_df) + 1
    )
)

combination_summary_df.to_csv(
    OUTPUT_DIR / "combination_summary.csv",
    index=False
)


# ============================================================
# 27. SAVE BEST OOF SUBMISSION
# ============================================================

best_row = combination_summary_df.iloc[0]

best_combo_id = best_row["combination_id"]

best_combo_idx = int(
    best_row["combination_index"]
)

best_submission_path = (
    SUBMISSION_DIR /
    "submission_BEST_BY_OOF_AUC.csv"
)

best_submission = pd.DataFrame({

    ID_COL: test_ids,

    TARGET_COL: combo_test_prediction_matrix[
        best_combo_idx
    ]

})

best_submission.to_csv(
    best_submission_path,
    index=False
)


# Also save a specific baseline alias.
baseline_submission_path = (
    SUBMISSION_DIR /
    "submission_B_BASELINE_CREDIT_TERM_ONLY.csv"
)

baseline_submission = pd.DataFrame({

    ID_COL: test_ids,

    TARGET_COL: combo_test_prediction_matrix[0]

})

baseline_submission.to_csv(
    baseline_submission_path,
    index=False
)


# ============================================================
# 28. FINAL SUBMISSION VALIDATION
# ============================================================

print("\nValidating all 64 submissions...")

submission_files = list(
    SUBMISSION_DIR.glob("submission_C*.csv")
)

if len(submission_files) != EXPECTED_COMBINATIONS:
    raise RuntimeError(
        f"Expected 64 combination submissions, "
        f"found {len(submission_files)}."
    )


for submission_path in submission_files:

    check = pd.read_csv(
        submission_path
    )

    if list(check.columns) != [
        ID_COL,
        TARGET_COL
    ]:
        raise RuntimeError(
            f"Invalid columns: {submission_path.name}"
        )

    if len(check) != N_TEST:
        raise RuntimeError(
            f"Invalid number of rows: {submission_path.name}"
        )

    if not check[ID_COL].equals(
        pd.Series(test_ids, name=ID_COL)
    ):
        raise RuntimeError(
            f"ID order mismatch: {submission_path.name}"
        )

    probabilities = check[TARGET_COL].to_numpy()

    if (
        not np.isfinite(probabilities).all()
        or (probabilities < 0).any()
        or (probabilities > 1).any()
    ):
        raise RuntimeError(
            f"Invalid probabilities: {submission_path.name}"
        )


# Validate best submission.
best_check = pd.read_csv(
    best_submission_path
)

if best_check.shape != (N_TEST, 2):
    raise RuntimeError(
        "Best submission validation failed."
    )


# ============================================================
# 29. SAVE FINAL EXPERIMENT METADATA
# ============================================================

final_metadata = {

    "experiment": "B_PLUS_ALL_COMBINATIONS_OF_SIX_FEATURES",

    "number_of_combinations": EXPECTED_COMBINATIONS,

    "number_of_models": EXPECTED_TOTAL_MODELS,

    "models_per_combination": B_MODEL_COUNT,

    "base_features": BASE_FEATURES,

    "fixed_feature": "CREDIT_TERM",

    "optional_features": NEW_FEATURES,

    "best_combination_id": str(best_combo_id),

    "best_optional_features": best_row["optional_features"],

    "best_pooled_oof_auc": float(
        best_row["pooled_oof_auc"]
    ),

    "best_mean_fold_auc": float(
        best_row["mean_fold_auc"]
    ),

    "maximum_iterations": MAX_ITERATIONS,

    "early_stopping_rounds": EARLY_STOPPING_ROUNDS,

    "best_model_selection": (
        "Highest pooled OOF ROC-AUC after averaging "
        "the two repeated validation predictions for every row."
    ),

    "all_models_saved": True,

    "cbm_and_json_saved_for_each_model": True,

    "test_predictions_saved_for_blending": True,

    "root_vectors_saved": True,

    "root_split_audits_saved": True,

    "test_prediction_matrix_shape": [
        EXPECTED_TOTAL_MODELS,
        N_TEST
    ],

    "root_vector_padding_value": 255,

    "baseline_experiment_directory": str(
        BASE_EXPERIMENT_DIR
    ),

    "output_directory": str(
        OUTPUT_DIR
    ),

    "submission_directory": str(
        SUBMISSION_DIR
    )

}


save_json(
    OUTPUT_DIR / "final_experiment_metadata.json",
    final_metadata
)


# ============================================================
# 30. FINAL REPORT
# ============================================================

print("\n" + "=" * 90)

print("B + SIX FEATURES: COMPLETE SEARCH FINISHED")

print("=" * 90)

print("Combinations:", len(combination_specs))

print("Models:", len(model_order_df))

print("Saved CBM models:", len(list(MODEL_DIR.rglob("*.cbm"))))

print("Saved JSON models:", len(list(MODEL_DIR.rglob("*.json"))))

print(
    "Saved individual root vectors:",
    len(list(VECTOR_INDIVIDUAL_DIR.rglob("*_root.npy")))
)

print("Test prediction matrix:", test_prediction_matrix.shape)

print("Padded root vector matrix:", padded_root_vectors.shape)

print("Combination prediction matrix:", combo_test_prediction_matrix.shape)

print("\nBEST COMBINATION")

print("Combination ID:", best_combo_id)

print("Features:", best_row["optional_features"])

print("Pooled OOF ROC-AUC:", f"{best_row['pooled_oof_auc']:.6f}")

print("Mean fold ROC-AUC:", f"{best_row['mean_fold_auc']:.6f}")

print("Mean tree count:", f"{best_row['mean_tree_count']:.1f}")

print("\nTOP 15 COMBINATIONS")

display_columns = [

    "rank",

    "combination_id",

    "optional_features",

    "pooled_oof_auc",

    "mean_fold_auc",

    "std_fold_auc",

    "pooled_oof_average_precision",

    "mean_tree_count"

]

print(
    combination_summary_df[
        display_columns
    ].head(15).to_string(index=False)
)


print("\nIMPORTANT OUTPUT FILES")

print("\nModels:")
print(MODEL_DIR)

print("\nAll test predictions for future blending:")
print(OUTPUT_DIR / "test_predictions.npy")

print("\nModel order:")
print(OUTPUT_DIR / "model_order.csv")

print("\nCombination test predictions:")
print(OUTPUT_DIR / "combination_test_predictions.npy")

print("\nRoot vectors:")
print(VECTOR_DIR / "root_vectors_padded.npy")

print("\nRoot vector summary:")
print(VECTOR_DIR / "root_vector_summary.csv")

print("\nFull root split audit:")
print(combined_audit_path)

print("\nPer-model validation metrics:")
print(OUTPUT_DIR / "per_model_validation_metrics.csv")

print("\nCombination summary:")
print(OUTPUT_DIR / "combination_summary.csv")

print("\nAll competition submissions:")
print(SUBMISSION_DIR)

print("\nBEST SUBMISSION:")
print(best_submission_path)

print("\nBASELINE SUBMISSION:")
print(baseline_submission_path)

print("\n" + "=" * 90)
print("ALL MODELS, ROOT VECTORS, PREDICTIONS AND SUBMISSIONS SAVED")
print("=" * 90)
```
